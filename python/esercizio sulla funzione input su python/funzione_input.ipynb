```python
#Esercizio 18: Input di una Data Semplice
#Chiedi all'utente di inserire il giorno, il mese e l'anno come numeri separati. Stampa la data nel formato "DD/MM/AAAA".

while True:
    giorno_input = input('Inserisci il giorno (1-31): ')
    if giorno_input.isdigit():
        giorno = int(giorno_input)
        if 1 <= giorno <= 31:
            break
    print('Giorno non valido. Inserisci un numero tra 1 e 31.')

while True:
    mese_input = input('Inserisci il mese (1-12): ')
    if mese_input.isdigit():
        mese = int(mese_input)
        if 1 <= mese <= 12:
            break
    print('Mese non valido. Inserisci un numero tra 1 e 12.')

while True:
    anno_input = input('Inserisci l\'anno (1900-2025): ')
    if anno_input.isdigit():
        anno = int(anno_input)
        if 1900 <= anno <= 2025:
            break
    print('Anno non valido. Inserisci un numero tra 1900 e 2025.')

print(f'La data inserita è: {giorno:02d}/{mese:02d}/{anno}')
```

```python
#Esercizio 23: Calcolo dell'Indice di Massa Corporea (BMI) Semplificato
#Chiedi all'utente il suo peso (in kg) e la sua altezza (in metri). Calcola un BMI semplificato (peso / altezza). Non è richiesta la radice quadrata.

peso = float(input('inserisci il tuo peso: '))
altezza = float(input('inserisci la tua altezza: '))

BMI_semplificato = peso / altezza
BMI_complesso = peso / (altezza * altezza)

print(f'BMI semplificato senza la radice quadrata: {BMI_semplificato:.2f}')
print(f'BMI complesso con la radice quadrata: {BMI_complesso:.2f}')
```
