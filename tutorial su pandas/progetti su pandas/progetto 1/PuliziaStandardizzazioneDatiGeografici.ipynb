Progetto 1: Pulizia e Standardizzazione di Dati Geografici 🗺️
Obiettivo: Importare un dataset contenente informazioni 
geografiche disordinate e prepararlo per un'analisi o 
una visualizzazione, focalizzandosi sulla pulizia e 
l'uniformità dei formati.

Fasi Procedurali Suggerite
Caricamento dei Dati:

Carica un file CSV (o un altro formato a tua scelta, 
ad esempio un dataset su città, stati o codici postali che puoi 
trovare online) in un DataFrame pandas.

Ispezione Iniziale:

Verifica i tipi di dati (.dtypes), il numero di valori mancanti 
(.isnull().sum()) e le prime righe (.head()).

Gestione dei Valori Mancanti:

Identifica le colonne con valori mancanti 
(ad es., latitudine o longitudine).

Decidi se eliminare le righe che contengono 
troppi NaN in colonne critiche o se imputare i valori mancanti 
(ad es., con la moda per i nomi delle regioni o lo zero per le 
popolazioni sconosciute).

Pulizia delle Stringhe:

Nella colonna dei nomi delle città/regioni:

Rimuovi gli spazi bianchi iniziali e finali (.str.strip()).

Converti tutti i nomi in minuscolo 
(.str.lower()) o in maiuscolo per uniformità.

Rimuovi caratteri speciali, punteggiatura non necessaria 
o numeri all'interno dei nomi.

Standardizzazione dei Formati Numerici:

Assicurati che le colonne come Popolazione o Area siano 
trattate come tipi numerici.

Se ci sono separatori di migliaia non standard 
(ad es., punti al posto delle virgole o viceversa), 
rimuovili o sostituiscili prima della conversione a 
numerico (pd.to_numeric()).

Creazione di Nuove Caratteristiche:

Crea una nuova colonna che categorizzi le città in 
base alla loro popolazione 
(ad es., 'Piccola', 'Media', 'Grande') 
utilizzando pd.cut() o np.select.

```python
import pandas as pd 

df = pd.read_csv('GeoInfo_Cittòvalle.csv')
```

```python
new_df = df.dropna()
new_df
```

```python
import numpy as np
```

```python
new_df['Latitudine'] = new_df['Latitudine'].replace(['nd', 'ND', 'N.D', 'N.D.'], np.nan)
new_df['Latitudine'] = pd.to_numeric(new_df['Latitudine'], errors='coerce')
new_df['Latitudine'].fillna(130, inplace = True)
new_df['Latitudine']
```

```python
new_df['Longitudine']
```

```python
new_df['Longitudine'] = new_df['Longitudine'].replace('-', np.nan)
new_df['Longitudine'] = pd.to_numeric(new_df['Longitudine'], errors='coerce')
new_df['Longitudine'].fillna(130, inplace = True)
new_df['Longitudine']
```

```python
new_df['Popolazione']
```

```python
new_df['Popolazione'] = new_df['Popolazione'].replace(['nd', '-', 0, -200, 'N.D.'], np.nan)
new_df['Popolazione'] = pd.to_numeric(new_df['Popolazione'], errors='coerce')
new_df['Popolazione'].fillna(130, inplace = True)
new_df['Popolazione']
```

```python
new_df['Altitudine (m)']
```

```python
new_df['Altitudine (m)'] = new_df['Altitudine (m)'].replace(['-', 'N.D.'], np.nan)
new_df['Altitudine (m)'] = pd.to_numeric(new_df['Altitudine (m)'], errors='coerce')
new_df['Altitudine (m)'].fillna(130, inplace = True)
new_df['Altitudine (m)']
```

```python
new_df['Superficie (km2)']
```

```python
new_df['Superficie (km2)'] = new_df['Superficie (km2)'].replace('n/d', np.nan)
new_df['Superficie (km2)'] = pd.to_numeric(new_df['Superficie (km2)'], errors='coerce')
new_df['Superficie (km2)'].fillna(130, inplace = True)
new_df['Superficie (km2)']
```
