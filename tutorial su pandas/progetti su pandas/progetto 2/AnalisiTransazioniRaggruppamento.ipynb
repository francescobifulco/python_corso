Obiettivo: Analizzare un dataset di transazioni di un e-commerce per calcolare le metriche chiave come il valore medio degli ordini, le vendite totali per prodotto e il cliente più fedele.

Fasi Procedurali Suggerite
Caricamento e Preparazione delle Colonne:

Carica un dataset di transazioni (dovrebbe includere colonne come ID Transazione, ID Cliente, Data, Importo, Prodotto).

Assicurati che la colonna Data sia di tipo datetime (pd.to_datetime()).

Assicurati che la colonna Importo sia numerica.

Analisi Temporale:

Imposta la colonna Data come indice del DataFrame (.set_index()).

Calcola le vendite totali mensili utilizzando il metodo resample() con l'aggregazione sum().

Analisi per Prodotto:

Utilizza la funzione groupby() sulla colonna Prodotto.

Calcola:

Le vendite totali per ogni prodotto (.sum() sull'Importo).

Il numero di transazioni per ogni prodotto (.size() o .count()).

Analisi per Cliente:

Utilizza groupby() sulla colonna ID Cliente.

Calcola:

La spesa totale di ogni cliente.

Il numero di ordini effettuati da ogni cliente.

L'importo medio speso per ordine da ogni cliente.

Ordina il risultato per identificare il cliente con la spesa totale più alta (.sort_values()).

Unione di Dati (Opzionale):

Se hai un dataset separato con i dettagli anagrafici dei clienti (ID Cliente, Nome, Paese), unisci i due DataFrame utilizzando pd.merge() per aggiungere le informazioni anagrafiche al riepilogo delle spese.

```python
import pandas as pd

df = pd.read_csv('E-commerce.csv')
df
```

```python
df.dtypes
```

```python
df['Data'].dtype
```

```python
df['Data'] = pd.to_datetime(df['Data'])
df['Data'].dtype
```

```python
df['Importo (€)'].dtype
```

```python
df = df.set_index('Data')
df
```

```python
vendite = df['Importo (€)'].resample('M').sum()
vendite
```

```python
vendite_pro = df.groupby('Prodotto')['Importo (€)'].sum()
vendite_pro
```

```python
transazioni_pro = df.groupby('Prodotto')['ID Transazione'].count()
transazioni_pro
```

```python
spesa_cliente = df.groupby('ID Cliente')['Importo (€)'].sum()
spesa_cliente
```

```python
ordini_cliente = df.groupby('ID Cliente')['ID Transazione'].count()
ordini_cliente
```

```python
importo_cliente = df.groupby('ID Cliente')['Importo (€)'].mean()
importo_cliente
```

```python
spesa_alta = df.sort_values(by='Importo (€)', ascending=False)
spesa_alta
```

```python
df1 = pd.read_csv('Dati_Anagrafici.csv')
df1
```

```python
df_merge = pd.merge(df, df1)
df_merge
```

```python
df_merge.to_csv('E-commerce_nuovo.csv', index=False)
```
