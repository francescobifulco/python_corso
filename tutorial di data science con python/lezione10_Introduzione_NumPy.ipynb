```python
import numpy as np
```

```python
uno = np.array([2, 6, 8])
due = np.array([9, 5, 6])
```

```python
uno, due
```

```python
print(f'Somma: {uno + due}')
```

```python
print(f'Prodotto: {uno * due}')
```

```python
print(f'Elevazione al quadrato: {uno ** 2}')
```

```python
print(f'Prodotto scalare: {np.dot(uno, due)}')
```

```python
# Stipendi
salario_gio = [3000, 2800, 2600, 3000, 2500]
salario_val = [3300, 2900, 2800, 3100, 2320]
salario_mar = [8500, 1800, 6600, 8000, 2530]
```

```python
riep_salari = np.array([salario_gio, salario_mar, salario_val])
riep_salari
```

```python
# Bonus
bonus_gio = [100, 800, 600, 200, 500]
bonus_val = [300, 900, 800, 100, 320]
bonus_mar = [500, 800, 600, 200, 530]
```

```python
riep_bonus = np.array([bonus_gio, bonus_mar, bonus_val])
riep_bonus
```

```python
# Operazioni semplici tra matrici np, +
riepilogo = riep_salari + riep_bonus
riepilogo
```

```python
# Operazioni semplici tra matrici np, *
riepilogo = riep_salari * riep_bonus
riepilogo
```

```python
type(riepilogo)
```

```python
# Funzioni statistiche
riepilogo.max()
```

```python
np.amax(riepilogo, axis=0)
```

```python
np.amax(riepilogo, axis=1)
```
