```python
import spacy
```

```python
nlp = spacy.load('it_core_news_sm')
```

```python
testo = '''Il gatto soriano, con la coda a punto interrogativo, 
balzò dal davanzale della finestra atterrando con 
grazia felina sul tappeto persiano. Fuori, 
il vento fischiava tra le foglie degli alberi di tiglio, 
portando con sé l'odore salmastro di un mare lontano. 
Una vecchia radio gracchiava una melodia jazz dimenticata, 
mentre un libro con la copertina sgualcita giaceva aperto, 
la pagina ingiallita segnata da un paio di occhiali da lettura. 
In lontananza, si sentiva il clacson isolato di un tram. 
L'universo, in quel preciso istante, 
sembrava contenuto nel ronzio di un frigorifero e nel 
ticchettio costante dell'orologio a pendolo'''
```

```python
estrazione = nlp(testo)
```

```python
# Creo un dizionario con le informazioni linguistiche estratte
analisi = {}

for token in estrazione:
    analisi[token.text] = {
        'pos': token.pos_,
        'dep': token.dep_   
    }

#print(analisi)
for parola, info in analisi.items():
    print(f'{parola:10} - POS: {info['pos']:10} DEP: {info['dep']:10}')
```

```python
# Analisi dei sostantivi
sostantivi = {k: v for k, v in analisi.items() if v['pos'] in ('NOUN', 'PROPN')}
```

```python
for parola, info in sostantivi.items():
    print(f'{parola:10} - POS: {info['pos']:10} DEP: {info['dep']:10}')
```
