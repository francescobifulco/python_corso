```python
import spacy
import json
```

```python
# Caricamento del modello
nlp = spacy.load('it_core_news_sm')
```

```python
# Leggi il contenuto del file TXT
with open('testo.txt', encoding='utf-8') as file:
    testo = file.read()
```

```python
# Analisi del testo
estrazione = nlp(testo)
```

```python
risultato = []
for token in estrazione:
    risultato.append(
        {
            'text': token.text,
            'pos': token.pos_,
            'dep': token.dep_,
            'lemma': token.lemma_
        }
    )
risultato
```

```python
with open('analisi.json', 'w', encoding='utf-8') as file:
    json.dump(risultato, file, indent=2, ensure_ascii=False)

print('Operazione effettuata con successo!')
    
```

```python
# Estrazione sostantivi
sostantivi = [
    {
            'text': token.text,
            'pos': token.pos_,
            'dep': token.dep_,
            'lemma': token.lemma_
        } for token in estrazione if token.pos_ in ("NOUN", 'PROPN')
]
```

```python
with open('sostantivi.json', 'w', encoding='utf-8') as file1:
    json.dump(sostantivi, file1, indent=4, ensure_ascii=False)

print('Operazione effettuata con successo!')
```
