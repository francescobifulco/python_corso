```python
df.sort_values(by='Età', inplace=True)
```

```python
bins = df['Età']
```

```python
plt.hist(df['Età'], bins=bins, color='skyblue', edgecolor='black')
plt.xlabel('Età')
plt.ylabel('Frequenza')
plt.title('Distribuzione delle Età')
plt.show()
```

```python
sns.boxplot(x='Età', y='Città', data=df)
plt.xlabel('Età')
plt.ylabel('Città')
plt.title("Box Plot: Totali per Punteggio")
plt.show()
```

```python
x = df['Età']
y = df['Punteggio']
```

```python
plt.scatter(x, y, c='red', marker='o', cmap='Greens')
plt.xlabel('Età')
plt.ylabel('Punteggio')
plt.title('')
plt.show()
```

```python
dati = {
    'Eta': df['Età'],
    'Punteggio': df['Punteggio'],
    'Reddito': df['Reddito']
}
```

```python
df_corr = pd.DataFrame(dati)
```

```python
corr_matrix = df_corr.corr()
corr_matrix
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Matrice di Correlazione')
plt.show()
```
