```python
tabella = {
    'Nome': ['Giovanni', 'Valeria', 'Marika'],
    'Eta': [38, 19, 29],
    'Citta': ['Napoli', 'Pescara', 'Chieti'],
    'Lavoro': ['Cuoco', np.nan, 'Segretaria']
}
```

```python
df = pd.DataFrame(tabella)
df
```

```python
# Operazioni semplici
df['Nome']
```

```python
df['Nome'][0] # Accedo prima alla colonna e poi alla riga!
```

```python
df.iloc[0] # Restituisci la riga all'indice 0
```

```python
df.loc[1, 'Citta'] # Accesso alla cella specifica in modalita esplicita
```

```python
# Oerazione utile
df.describe() # Accesso alla statistiche
```

```python
df.info() # Statistiche sui dati
```
