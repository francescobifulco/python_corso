```python
import pandas as pd
```

```python
# Seria con indice automatico
persona = ['Giovanni', 'Valeria', 'Mario']
```

```python
invitati = pd.Series(persona)
invitati
```

```python
# Serie con indice personalizzato
numeri = pd.Series([5,6,98,6,354])
numeri
```

```python
numeri = pd.Series([5,6,98,6,354], index=['a', 'b', 'c', 'd', 'e'])
numeri
```

```python
numeri_mist = pd.Series([5,6,98,6,354, 'franco'], index=['a', 'b', 'c', 'd', 'e', 'f'])
numeri_mist
```
