```python
# Unpacking
persona = ('Franco', 'Rossi', 34, True)
nome, cognome, eta, hasPatente = persona
```

```python
print(f'Nome: {nome}')
print(f'Cognome: {cognome}')
print(f'Eta: {eta}')
print(f'Ha la patente: {hasPatente}')
```

```python
# Tupla come tipo di ritorno
def dettaglioPersona():
    return ('Franco', 'Rossi', 34, True)

nome, cognome, eta, hasPatente = dettaglioPersona()

print(f'Nome: {nome}')
print(f'Cognome: {cognome}')
print(f'Eta: {eta}')
print(f'Ha la patente: {hasPatente}')
```
