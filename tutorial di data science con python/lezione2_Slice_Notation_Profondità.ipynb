```python
# Utilizzo: lista[star:stop:step]
# start: indice di partenza
# stop: indice finale
# step: numero di salti da fare
```

```python
lista = ['Giovanni', 'Valeria', 'Marika', 'Mario', 'Ester']
lista
```

```python
# dal secondo al terso elemento escludendo il quarto
lista[1:3]
```

```python
# Solo i primi 3 elementi
lista[:3]
```

```python
# Dal secondo elemento in poi
lista[2:]
```

```python
# Prende gli ultimi due elementi
lista[-2:]
```

```python
# Tutta la lista tranne ultimo elemento
lista[:-1]
```

```python
# Un elemento ogni due
lista[::2]
```

```python
# Stampa al contrario della lista
lista[::-1]
```

```python
lista[1:4:2]
```
