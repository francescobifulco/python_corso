```python
coda =[]
```

```python
# Aggiunta in fondo (ENQUEUE)
coda.append('Giovanni')
coda.append('Valeria')
coda.append('Marika')
coda.append('Mario')
coda.append('Ester')
```

```python
coda
```

```python
# Rimozione dal fronte (ENQUEUE)
persona = coda.pop(0)
print(f'Estratto: {persona}')
```

```python
coda
```

```python
from collections import deque
```

```python
coda = deque()

# (ENQUEUE)
coda.append('Giovanni')
coda.append('Valeria')
coda.append('Marika')
coda.append('Mario')
coda.append('Ester')

coda
```

```python
# (DENQUEUE)
persona = coda.popleft()
print(f'Estratto: {persona}')
```

```python
coda
```

```python
# Svuota la coda
coda.clear()
coda
```

```python
# Aggiunge un elemento nella coda
coda.appendleft('Ester')
coda
```
