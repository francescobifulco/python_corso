```python
# Importa la libreria spaCy
import spacy
```

```python
# Carica il modello linguistico italiano di piccole dimensioni 
# ('it_core_news_sm')
# Questo modello è pre-addestrato per l'elaborazione del 
# linguaggio naturale (NLP) in italiano
nlp = spacy.load('it_core_news_sm')
```

```python
# Definisce la stringa di testo su cui eseguire l'analisi NLP
testo = '''Il gatto soriano, con la coda a punto interrogativo, 
balzò dal davanzale della finestra atterrando con 
grazia felina sul tappeto persiano. Fuori, 
il vento fischiava tra le foglie degli alberi di tiglio, 
portando con sé l'odore salmastro di un mare lontano. 
Una vecchia radio gracchiava una melodia jazz dimenticata, 
mentre un libro con la copertina sgualcita giaceva aperto, 
la pagina ingiallita segnata da un paio di occhiali da lettura. 
In lontananza, si sentiva il clacson isolato di un tram. 
L'universo, in quel preciso istante, 
sembrava contenuto nel ronzio di un frigorifero e nel 
ticchettio costante dell'orologio a pendolo'''
```

```python
# Esegue l'elaborazione del testo con spaCy
# 'estazione' è un oggetto Doc di spaCy, 
# che contiene i token e i risultati dell'analisi
estazione = nlp(testo)
estazione
```

```python
# Itera su ciascun token nell'oggetto Doc e stampa la sua:
# 1. Forma testuale (token.text)
# 2. Part-of-Speech (POS) Tag, ovvero la categoria grammaticale (token.pos_)
# 3. Dipendenza sintattica (token.dep_)
for token in estazione:
    print(token.text, token.pos_, token.dep_)
```

```python
stack = [] #  Pila contenente tutti i token soggetto
```

```python
for token in estazione: # Popola la pila con solo i token di tipo NPUN e PROPN
    # Verifica se la categoria grammaticale (POS) del token è 
    # un Sostantivo ('NOUN') o un Nome Proprio ('PROPN')
    if token.pos_ == 'NOUN' or token.pos_ == 'PROPN':
        # Se la condizione è vera, aggiunge la 
        # forma testuale del token alla pila
        stack.append(token.text)

# Controlla se la pila non è vuota
if stack:
    # Inizializza una stringa vuota per costruire l'output finale
    output = ''
    # Finché la pila non è vuota, esegue il seguente blocco
    while stack:
        # Estrae e rimuove l'ultimo elemento aggiunto alla pila 
        # (operazione LIFO: Last-In, First-Out)
        # Questo elemento (sostantivo/nome proprio) viene poi 
        # prepeso alla stringa 'output'
        # L'effetto è che i sostantivi vengono stampati 
        # nell'ordine INVERSO rispetto a come sono stati trovati nel testo
        output = stack.pop() + ' ' + output
    # Stampa la stringa finale contenente tutti i 
    # sostantivi/nomi propri in ordine inverso (separati da uno spazio)
    print(output)
```
