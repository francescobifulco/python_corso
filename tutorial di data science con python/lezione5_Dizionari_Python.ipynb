```python
persona = {
    'nome': 'Valerio',
    'cognome': 'Esposito',
    'eta': 40,
    'hasPatente': True
}

persona
```

```python
# Accesso
persona['nome']
```

```python
# Modifica
persona['nome'] = 'Marco'
persona['nome']
```

```python
nominativo = persona['nome']
nominativo
```

```python
numero_animali = persona.get('animali', 'Non specificato')
numero_animali
```

```python
# Eliminazione
del persona['hasPatente']
persona
```

```python
# Ottenere tutte le chiavi
chiavi = persona.keys()
chiavi
```

```python
# Ottenere tutti i valori
valori = persona.values()
valori
```

```python
# Ottenere tutti gli eleminti chiave-valori
elementi = persona.items()
elementi
```

```python
# Aggiornamento dei valori delle chiavi
aggiorna = {
    'nome': 'Mario',
    'cognome': 'Mario',
    'eta': 55,
    'hasPatente': False
}
persona.update(aggiorna)
persona
```

```python
# Elimina tutte le combinazioni del dictionary
persona.clear()
persona
```

```python
persona = {
    'nome': 'Valerio',
    'cognome': 'Esposito',
    'eta': 40,
    'hasPatente': True
}

persona
```

```python
# Accesso interpolato agli elementi
chiavi = persona.keys()
for item in chiavi:
    print(f'Chiavi: {item}')

valori = persona.values()
for item in valori:
    print(f'Valori: {item}')

elementi = persona.items()
for chiavi,valori in elementi:
    print(f'Chiavi {chiavi}: valori {valori}')
```

```python
from collections import OrderedDict
```

```python
dizionario = {
    'd': 4,
    'a': 1,
    'b': 3,
    'c': 2
}
```

```python
# Ordino rispetto alle chiavi
diz_ord_chiav = OrderedDict(sorted(dizionario.items()))
diz_ord_chiav
```

```python
# Ordino rispetto alle valori
diz_ord_valore = OrderedDict(sorted(dizionario.items(), key=lambda x: x[1]))
diz_ord_valore
```
