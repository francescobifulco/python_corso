```python
import mysql.connector
```

```python

mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="mydatabase"
)

mycursor = mydb.cursor()
```

```python
sql = "DELETE FROM cliente WHERE indirizzo = 'Corso Umberto I, 120 - Bari'"

mycursor.execute(sql)

mydb.commit()

print(mycursor.rowcount, "record(s) deleted")
```

```python
sql = "DELETE FROM cliente WHERE indirizzo = %s"
adr = ("Piazza della Repubblica, 11 - Siena", )

mycursor.execute(sql, adr)

mydb.commit()

print(mycursor.rowcount, "record(s) deleted") 
```
