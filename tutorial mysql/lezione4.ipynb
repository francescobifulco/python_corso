```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="mydatabase"
)

mycursor = mydb.cursor()
```

```python

sql = "INSERT INTO cliente (nome, cognome, indirizzo) VALUES (%s, %s, %s)"
val = ("John", "Highway", "Via del Corso, 12 - Roma")
mycursor.execute(sql, val)

mydb.commit()

print(mycursor.rowcount, "record inserted.")
```
