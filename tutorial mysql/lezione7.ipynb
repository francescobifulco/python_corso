```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="mydatabase"
)

mycursor = mydb.cursor()
```

```python
sql = "SELECT * FROM cliente ORDER BY nome"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
  print(x)
```

```python

sql = "SELECT * FROM cliente ORDER BY nome DESC"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
  print(x) 
```
