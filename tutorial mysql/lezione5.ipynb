```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="mydatabase"
)

mycursor = mydb.cursor()
```

```python
mycursor.execute("SELECT * FROM cliente")

myresult = mycursor.fetchall()

for x in myresult:
  print(x)
```
