## Progetto 1.1: Linee e Punti Semplici
Task: Traccia un grafico a linea per una semplice funzione matematica (es. $y = x^2$) e uno scatter plot (diagramma a dispersione) per dati casuali. Aggiungi titoli e etichette agli assi.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(10)
y = x**2
y
```

```python
plt.plot(x,y, label='y = x^2', color='blue')
plt.title('Linee e Punti Semplici')
plt.xlabel('valori della X')
plt.ylabel('valori della y')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.scatter(x,y, c=y,cmap='Greens')
cbar = plt.colorbar()
cbar.set_label('Intensita')
plt.title('Grafico dispersione')
plt.xlabel('valori della X')
plt.ylabel('valori della Y')
plt.grid(True)
plt.show()
```
