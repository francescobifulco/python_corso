## Istogrammi e Grafici a Barre
1. Crea un istogramma dei voti degli studenti (dati casuali) per visualizzare la distribuzione delle frequenze.

2. Crea un grafico a barre che mostri le vendite mensili di 5 prodotti.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.random.randint(0,11, size=15)
x
```

```python
plt.hist(x, bins=range(0, 12), edgecolor='black', color='skyblue', align='left')
plt.xlabel('Valori')
plt.ylabel('Frequenza')
plt.title('Distribuzione dei voti degli studenti')
plt.xticks(range(0, 11))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
prodotti = ['TV', 'PC', 'Tavolo', 'Sedia', 'Specchi']
vendite = [110, 200, 150, 70, 300]
```

```python
plt.bar(prodotti, vendite, color='lightgreen', edgecolor='black')
plt.xlabel('Prodotti')
plt.ylabel('Vendite mensile')
plt.title('Le vendite mensile dei prodotti')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
