## Grafico 3D (Superficie o Dispersione)
1. Genera dati per una superficie 3D (es. la funzione $z = \sin(\sqrt{x^2 + y^2})$) e visualizzala usando ax.plot_surface().

2. Aggiungi una barra dei colori (fig.colorbar()) per rappresentare i valori di $z$.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D 
```

```python
x = np.linspace(-10, 10, 200)
y = np.linspace(-10, 10, 200)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='plasma', edgecolor='none', alpha=0.9)

fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
ax.set_title("Superficie 3D")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```
