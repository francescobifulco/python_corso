## Mappe Geospaziali (Introduzione a Cartopy)

1. Seleziona un set di dati con coordinate di latitudine e longitudine (es. posizioni di terremoti).

2. Utilizza Cartopy per tracciare queste posizioni su una mappa del mondo o di una regione specifica.

```python
import matplotlib.pyplot as plt
import pandas  as pd
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

```python
df = pd.read_csv('terremoti.csv')
df
```

```python
plt.figure(figsize=(10,6))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.coastlines()
ax.add_feature(cfeature.BORDERS)
sc = ax.scatter(df['longitude'], df['latitude'], 
                c=df['magnitude'], cmap='hot', s=50, alpha=0.7,
                transform=ccrs.PlateCarree())
plt.colorbar(sc, label='Magnitudo')
plt.title('Mappa terremoti')

plt.show()
```
