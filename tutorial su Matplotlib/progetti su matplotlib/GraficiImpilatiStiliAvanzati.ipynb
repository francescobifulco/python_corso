## Grafici Impilati e Stili Avanzati
1. Crea un grafico a barre impilate che mostri i ricavi trimestrali di due prodotti (prodotto A e B) per un anno.

2. Sperimenta con diversi stili integrati di Matplotlib (es. 'ggplot', 'seaborn', 'dark_background') per cambiare rapidamente l'aspetto dei tuoi grafici.

```python
import matplotlib.pyplot as plt
```

```python
prodottoA = [25000, 30000, 28000, 35000]
prodottoB = [18000, 20000, 22500, 25000]
categorie = ['Q1', 'Q2', 'Q3', 'Q4']
```

```python
plt.bar(categorie, prodottoA, label='Prodotto A')
plt.bar(categorie, prodottoB, bottom=prodottoA, label='Prodotto B')
plt.xlabel('Trimestre')
plt.ylabel('Ricavi (€)')
plt.title('Ricavi trimestrali di due prodotti')
plt.legend()
plt.show()
```

```python
plt.style.use('ggplot')
plt.bar(categorie, prodottoA, label='Prodotto A')
plt.bar(categorie, prodottoB, bottom=prodottoA, label='Prodotto B')
plt.xlabel('Trimestre')
plt.ylabel('Ricavi (€)')
plt.title('Ricavi trimestrali di due prodotti')
plt.legend()
plt.show()
```

```python
plt.style.use('dark_background')
plt.bar(categorie, prodottoA, label='Prodotto A')
plt.bar(categorie, prodottoB, bottom=prodottoA, label='Prodotto B')
plt.xlabel('Trimestre')
plt.ylabel('Ricavi (€)')
plt.title('Ricavi trimestrali di due prodotti')
plt.legend()
plt.show()
```
