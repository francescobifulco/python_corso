## Grafico a Torta e Sottografici (Subplots)
1. Crea un grafico a torta che mostri la ripartizione del budget annuale in 4 categorie. Usa l'opzione autopct per visualizzare le percentuali.

2. Usa plt.subplots() per visualizzare il grafico a linea del Progetto 1.1 e l'istogramma del Progetto 1.2 nella stessa figura, uno accanto all'altro (1 riga, 2 colonne).

```python
import matplotlib.pyplot as plt
```

```python
percentuali = slices = [60,40,50,30]
categorie = ['telefoni', 'pc', 'modili', 'eletronica']
colori = ['red','brown', 'orange','green']
```

```python
plt.pie(percentuali, labels=categorie, colors=colori, wedgeprops={'edgecolor': '#000000'}, autopct='%1.1f%%')
plt.title('Ripartizione del budget annuale in 4 categorie')
plt.show()
```

```python
import numpy as np
```

```python
x1 = np.arange(10)
y = x1**2
y
```

```python
x = np.random.randint(0,11, size=15)
x
```

```python
fig, (ax1, ax2 ) = plt.subplots(nrows=1, ncols=2, sharex=False)
ax1.plot(x1,y, label='y = x^2', color='blue')
ax1.set_title('Linee e Punti Semplici')
ax1.set_xlabel('valori della X')
ax1.set_ylabel('valori della y')
ax1.grid(True)
ax1.legend()

ax2.hist(x, bins=range(0, 12), edgecolor='black', color='skyblue', align='left')
ax2.set_xlabel('Valori')
ax2.set_ylabel('Frequenza')
ax2.set_title('Distribuzione dei voti degli studenti')
ax2.set_xticks(range(0, 11))
ax2.grid(axis='y', linestyle='--', alpha=0.7)
```
