## Box Plot e Grafico a Violino (Statistica)
Genera tre set di dati casuali (simulando, ad esempio, tre diversi trattamenti medici) e confronta la loro distribuzione usando:

1. Un Box Plot.

2. Un Grafico a Violino.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
trattamentoA = [8.2, 7.5, 6.9, 9.1, 8.0, 7.8, 6.5, 9.3, 8.7, 7.1, 6.8, 8.4, 7.9, 8.1, 7.3]
trattamentoB = [12.5, 13.1, 14.3, 12.8, 13.9, 15.2, 13.7, 14.1, 12.4, 14.5, 13.3, 14.0, 13.6, 12.9, 13.8]
trattamentoC = [2.3, 1.9, 2.1, 3.0, 1.8, 2.4, 2.2, 2.0, 1.7, 2.5, 1.6, 2.6, 2.1, 1.9, 2.3]
```

```python
plt.boxplot([trattamentoA, trattamentoB, trattamentoC], labels=['Trattamento A', 'Trattamento B', 'Trattamento C'])
plt.title('Box Plot dei Trattamenti')
plt.ylabel('Valori dei trattamenti')
plt.xlabel('Trattamenti')
plt.grid(True)
#plt.tight_layout()
plt.show()
```

```python
data = [trattamentoA, trattamentoB, trattamentoC]
labels = ['Trattamento A', 'Trattamento B', 'Trattamento C']
```

```python
sns.violinplot(data=data)
plt.xticks([0, 1, 2], labels)
plt.title('Grafico a Violino dei Trattamenti')
plt.ylabel('Valori dei trattamenti')
plt.xlabel('Trattamenti')
plt.grid(True)
plt.show()
```
