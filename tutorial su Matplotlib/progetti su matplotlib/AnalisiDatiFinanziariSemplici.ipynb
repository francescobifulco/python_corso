## Analisi di Dati Finanziari Semplici
Simula i prezzi azionari nel tempo e traccia un grafico a linea che mostri il trend.

Imposta le etichette sull'asse X in formato data.

Aggiungi delle annotazioni (plt.annotate()) per evidenziare il massimo e il minimo.

```python
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
from matplotlib import dates as mpl_dates
```

```python
date = [
    datetime(2022,1,10),
    datetime(2022,2,15),
    datetime(2022,3,20),
    datetime(2022,4,25),
    datetime(2022,5,30),
    datetime(2022,7,10),
    datetime(2022,9,15),
    datetime(2022,12,20),
    datetime(2023,1,25),
    datetime(2023,10,30)
]
profitto = [100,140,105,200,322,400,200,120,110,110] 
```

```python
data_formatate =mpl_dates.DateFormatter('%d %b %Y')
```

```python
idx_max = profitto.index(max(profitto))
idx_min = profitto.index(min(profitto))

data_max = date[idx_max]
data_min = date[idx_min]

val_max = profitto[idx_max]
val_min = profitto[idx_min]
```

```python
plt.plot_date(date,profitto, linestyle='solid')
plt.gcf().autofmt_xdate()
plt.gca().xaxis.set_major_formatter(data_formatate)
plt.title('Analisi dei dati finanza')
plt.xlabel('data')
plt.ylabel('profitto')
plt.annotate('',
             xy=(data_max, val_max),
             xytext=(data_max, val_max + 20),
             arrowprops=dict(arrowstyle='->', color='green'),
             color='green')
plt.annotate('',
             xy=(data_min, val_min),
             xytext=(data_min, val_min - 30),
             arrowprops=dict(arrowstyle='->', color='red'),
             color='red')
plt.show()
```
