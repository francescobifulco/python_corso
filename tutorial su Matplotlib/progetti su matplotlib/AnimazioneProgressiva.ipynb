## Animazione Progressiva

1. Crea un semplice grafico a linea animato che simuli la crescita dei dati nel tempo, aggiungendo un nuovo punto ad ogni frame.

2. L'animazione deve aggiornare solo i dati del grafico, non ricrearlo interamente.

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
```

```python
tempo = np.arange(0,11)
valori1 = 2 * tempo
valori2 = 3 * tempo
valori3 = 4 * tempo
```

```python
fig, ax = plt.subplots()
plot1 = ax.plot(tempo, valori1)
plot2 = ax.plot(tempo, valori2)
plot3 = ax.plot(tempo, valori3)
ax.set_xlim(0, 10)
ax.set_ylim(0, 50)
line1, = ax.plot([], [], label='valori *2')
line2, = ax.plot([], [], label='valori *3')
line3, = ax.plot([], [], label='valori *4')
plt.legend()
```

```python
def update(frame):
    line1.set_data(tempo[:frame], valori1[:frame])
    line2.set_data(tempo[:frame], valori2[:frame])
    line3.set_data(tempo[:frame], valori3[:frame])
    return line1, line2, line3
```

```python
animazione = animation.FuncAnimation(
    fig,
    update,
    frames=len(tempo) + 1,
    interval=300,
    blit=True,
    repeat=False
)

plt.show()
```
