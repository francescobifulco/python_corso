```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
```

```python
fig, ax = plt.subplots()
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.2, 1.2)
line, = ax.plot([], [], lw=2)
```

```python
def init():
    line.set_data([], [])
    return line,

# Funzione di aggiornamento
def update(frame):
    line.set_data(x[:frame], y[:frame])
    return line,
```

```python
ani = animation.FuncAnimation(
    fig, update, frames=len(x), init_func=init,
    blit=True, interval=50, repeat=False
)

plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots()
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.5, 1.5)

line1, = ax.plot([], [], label='sin(x)')
line2, = ax.plot([], [], label='cos(x)')
plt.legend()

def init():
    line1.set_data([], [])
    line2.set_data([], [])
    return line1, line2

def update(frame):
    line1.set_data(x[:frame], y1[:frame])
    line2.set_data(x[:frame], y2[:frame])
    return line1, line2

ani = animation.FuncAnimation(
    fig, update, frames=len(x), init_func=init,
    blit=True, interval=50, repeat=False
)

plt.show()
```

```python
fig, ax = plt.subplots()
xdata, ydata = [], []
line, = ax.plot([], [], 'r-')
ax.set_xlim(0, 100)
ax.set_ylim(0, 10)

def update(frame):
    xdata.append(frame)
    ydata.append(np.random.rand() * 10)
    line.set_data(xdata, ydata)
    ax.set_xlim(0, max(100, frame + 10))  # aggiorna se serve
    return line,

ani = animation.FuncAnimation(
    fig, update, frames=range(200),
    blit=True, interval=100
)

plt.show()
```

```python
ani.save("animazione.mp4", writer='ffmpeg', fps=30)
```

```python
ani.save("animazione.gif", writer='pillow', fps=10)
```
