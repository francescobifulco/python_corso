```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips")
print(df.head())
```

```python
# Crea un Violin Plot per visualizzare la distribuzione di 'total_bill'
# (asse y) per ogni 'day' (asse x).
sns.violinplot(x="day", y="total_bill", data=df)
plt.title("Violin Plot: Total Bill per Giorno")
plt.show()
```
