```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
```

```python
# 1. SCATTER PLOT 3D (Punti Casuali)

# Genera 100 coordinate x, y, z casuali tra 0 e 1.
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()
# Aggiunge un subplot 3D alla figura. La proiezione '3d' è essenziale.
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, c='red', marker='o')

# Crea lo scatter plot 3D:
# I dati sono x, y, z. 'c' definisce il colore, 'marker' la forma del punto.
ax.set_title("Scatter 3D")
plt.show()
```

```python
# 2. LINE PLOT 3D (Curva Elicoidale)

# Crea un array di 100 punti temporali.
t = np.linspace(0, 10, 100)
# Definisce le coordinate per creare una 
# spirale o elica (sin(t), cos(t) è un cerchio, Z=t estende nello spazio).
x = np.sin(t)
y = np.cos(t)
z = t

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Traccia la curva nello spazio 3D.
ax.plot(x, y, z, label='Curva 3D')
ax.legend()

plt.show()
```

```python
# 3. SURFACE PLOT 3D (Superficie di una funzione sinusoidale)

# Crea 100 punti tra -5 e 5 per gli assi X e Y.
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
# meshgrid crea una griglia 2D di coordinate dai vettori x e y.
X, Y = np.meshgrid(x, y)

# Calcola la funzione Z = sin(sqrt(X^2 + Y^2)) per ogni punto della griglia.
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Disegna la superficie 3D:
# 'cmap' (colormap) definisce lo schema di colori.
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Aggiunge una barra dei colori per interpretare i valori Z.
fig.colorbar(surf)

plt.show()
```

```python
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)
```

```python
ax.contour3D(X, Y, Z, 50, cmap='binary')
```

```python
# 4. SUPERFICIE COMPLESSA con Wireframe e Colorbar (Funzione Sinc 2D)

x = np.linspace(-6, 6, 150)
y = np.linspace(-6, 6, 150)
X, Y = np.meshgrid(x, y)

# Funzione Sinc 2D (sin(r)/r), usata spesso per dimostrazioni 3D.
# '1e-6' aggiunto per prevenire la divisione per zero al centro (X=0, Y=0).
Z = np.sin(np.sqrt(X**2 + Y**2)) / (np.sqrt(X**2 + Y**2) + 1e-6)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 1. SURFACE PLOT: Disegna la superficie con colori.
surf = ax.plot_surface(X, Y, Z, cmap='coolwarm', edgecolor='none', alpha=0.9)

# 2. WIREFRAME: Sovrappone una griglia di fili per visualizzare meglio la geometria.
ax.plot_wireframe(X, Y, Z, color='black', linewidth=0.3, alpha=0.2)

# 3. COLORBAR: Aggiunge la legenda dei colori.
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

# Etichette e titolo
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
# Titolo specifico per l'ultimo grafico
ax.set_title('Superficie 3D con wireframe')

plt.show()
```
