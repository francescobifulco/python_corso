```python
# Importa i moduli standard Python per la gestione 
# di date e intervalli di tempo.
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
# Importa il modulo specifico per la gestione 
# e la formattazione delle date in Matplotlib.
from matplotlib import dates as mpl_dates
```

```python
x = [
    datetime(2022,1,10),
    datetime(2022,1,15),
    datetime(2022,1,20),
    datetime(2022,1,25),
    datetime(2022,1,30)
] # Lista di oggetti datetime usati per l'asse X.
y = [3,7,2,9,1] # Valori numerici per l'asse Y.
```

```python
# --- Formattazione delle Date ---
data_formatate =mpl_dates.DateFormatter('%d %b %Y')
# Crea un oggetto Formatter per definire il 
# formato di visualizzazione delle date sull'asse X:
# '%d': Giorno del mese con due cifre (es. 10)
# '%b': Mese in forma abbreviata (es. Gen)
# '%Y': Anno con quattro cifre (es. 2022)
```

```python
plt.plot_date(x,y, linestyle='solid')
plt.gcf().autofmt_xdate()

# 'plt.gcf()' (Get Current Figure) prende la figura corrente.
# '.autofmt_xdate()' ruota automaticamente 
# le etichette sull'asse X per evitare che si sovrappongano.
plt.gca().xaxis.set_major_formatter(data_formatate)
# 'plt.gca()' (Get Current Axes) prende l'asse corrente.
# '.xaxis' si riferisce all'asse X.
# '.set_major_formatter()' applica 
# l'oggetto 'data_formatate' 
# per definire l'aspetto finale delle etichette.

plt.title('nome del grafico')
plt.show()
```
