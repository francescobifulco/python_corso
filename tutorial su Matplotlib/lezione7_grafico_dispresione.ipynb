```python
import matplotlib.pyplot as plt
```

```python
x = [1,5,6,8,3,2,2,7,9,1]
y = [4,7,2,7,9,3,2,4,5,2]
```

```python
# Grafico 1: Personalizzazione di Base (Marker e Colore Statico)

plt.scatter(x,y, s=150, # Dimensione (size) statica di tutti i punti.
            c='red', # Colore (color) statico di tutti i punti.
            marker='X')  # Forma del marcatore
plt.title('Grafico 1: Marker Statico')
plt.show()
```

```python
# Grafico 2: Bordo, Spessore e Trasparenza

plt.scatter(x,y, s=150, # Dimensione.
            c='red',  # Colore di riempimento.
            edgecolor='purple', # Colore del bordo (edge).
            linewidth=30, # Spessore della linea del bordo (ridotto da 30 per una visualizzazione migliore).
            alpha=0.5) # Trasparenza (alpha): 0.0 è trasparente, 1.0 è opaco.
plt.title('Grafico 2: Bordo e Trasparenza')
plt.show()
```

```python
colors= [2,4,6,7,1,9,4,7,2,8] # Lista di valori usati per mappare il colore di ciascun punto (dati di intensità).
```

```python
# Grafico 3: Mappatura Colore (Color Mapping)

plt.scatter(x,y, s=150, c=colors, 
            cmap='Greens') # Mappa di colore da usare (Greens = sfumature di verde).
plt.title('Grafico 3: Mappatura Colore')
plt.show()
```

```python
# Grafico 4: Aggiunta della Colorbar (Legenda Colore)

plt.scatter(x,y, s=150, c=colors, 
            cmap='Greens') # Mappa di colore.

# Aggiunta della Colorbar:
cbar = plt.colorbar() # Crea e visualizza la barra di riferimento per la mappatura del colore.
cbar.set_label('Intensita') # Etichetta per la Colorbar.
plt.title('Grafico 4: Con Colorbar')
plt.show()
```

```python
sizes = [200,250,120,500,430,700,567,312,691,443] # Lista di valori per definire la dimensione (area) di ciascun punto.
```

```python
# Grafico 5: Mappatura sia per Dimensione (Size) che per Colore (Color)

plt.scatter(x,y, s=sizes, c=colors, # Colori dinamici.
            cmap='Greens')
cbar = plt.colorbar()
cbar.set_label('Intensita')
plt.title('Grafico 5: Mappatura per Dimensione e Colore')
plt.show()
```
