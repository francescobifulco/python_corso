```python
import matplotlib.pyplot as plt
```

```python
x = [26,27,28,29,30,33,37]
y = [80,150,170,130,140,190,200]
```

```python
# Grafico 1: Area Semplice Sotto la Curva


plt.plot(x,y)
plt.fill_between(x,y,  # Dati X e Y. La seconda Y è implicita come 0 (l'asse X).
                 color='blue') # Colore solido per l'area.
plt.title('Grafico 1: Area tra la Curva e l\'Asse X')
plt.show()
```

```python
media = 125 # Valore di riferimento (una media o soglia).
```

```python
# Grafico 2: Area Semplice con Trasparenza (Alpha)

plt.plot(x,y)
plt.fill_between(x,y, alpha=0.5) # Imposta la trasparenza (0.0 è invisibile, 1.0 è opaco). Il colore è di default blu.
plt.title('Grafico 2: Area con Trasparenza (Alpha)')
plt.show()
```

```python
# Grafico 3: Area tra la Curva e una Linea di Riferimento (Media)

plt.plot(x,y)
plt.fill_between(x,y, media, # Dati X, Curva Y, e Valore Y2 (la media 125).
                 alpha=0.2)  # Trasparenza.
plt.title('Grafico 3: Area tra la Curva e la Media')
plt.show()
```

```python
import numpy as np

x = np.array([26,27,28,29,30,33,37])
y = np.array([80,150,170,130,140,190,200])
```

```python
# Grafico 4: Area Selettiva con Condizione Logica ('Where')

plt.plot(x,y)
plt.fill_between(x,y, media, where=(y > 150),alpha=0.2, # Condizione logica: colora l'area SOLO dove i valori Y sono maggiori di 150.
                 interpolate=True)# Garantisce che l'area sia chiusa correttamente anche se la linea attraversa la soglia.
plt.title('Grafico 4: Area Selettiva Sopra la Soglia')
plt.show()
```
