```python
# Importa la libreria 'matplotlib.pyplot' e 
# le assegna l'alias 'plt' per comodità.
import matplotlib.pyplot as plt
```

```python
# Dati per l'asse X e Y
dati_x = [1, 2, 3] # Lista dei valori per l'asse delle ascisse (X).
dati_y = [5, 7, 4] # Lista dei valori per l'asse delle ordinate (Y).
```

```python
# Crea il grafico:
# 'plt.plot()' genera un grafico a linea, 
# dove il primo argomento (dati_x) 
# definisce i punti sull'asse X e il 
# secondo (dati_y) definisce i punti corrispondenti sull'asse Y.


plt.plot(dati_x, dati_y)

# Visualizza il grafico:
# 'plt.show()' mostra la finestra del grafico creato. 
plt.show()
```
