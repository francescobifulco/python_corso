```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# --- Dati per il grafico ---
x = [25, 26, 27, 28, 29, 30, 31]
y1 = [100, 120, 130, 140, 150, 145, 140]
y2 = [90,80,100,120,130,200,300]
```

```python
# --- Preparazione per il Grafico a Barre Verticali Affiancate ---

indexs = np.arange(7) # Crea un array NumPy di indici: [0, 1, 2, 3, 4, 5, 6]
print(indexs) # Stampa gli indici per verifica.  
```

```python
width = 0.3 # Definisce la larghezza di ogni singola barra
```

```python
# --- Primo Grafico: Barre Verticali Affiancate (`plt.bar`) ---

# Barre della Serie 1 (Italiani): posizionate sull'indice 
# di partenza (0, 1, 2, ...).
plt.bar(indexs, y1, 
        label='italiani', 
        width=width,  # Larghezza definita per le barre.
        color='red')

# Barre della Serie 2 (Francesi): spostate a destra di 'width' per affiancarle.
plt.bar(indexs+width, y2, label='francesi', width=width, color='#212121') 

plt.title('nome del grafico') 
plt.xlabel('eta')
plt.ylabel('valori caso')
plt.legend()

# Imposta le etichette sull'asse X:
# La posizione è spostata di metà larghezza (width/2) 
# tra le due barre per centrare l'etichetta 'eta'.
# Le etichette sono i valori della lista 'x' (25, 26, ...).
plt.xticks(indexs+width/2, x)
plt.show()
```

```python
# --- Secondo Grafico: Barre Orizzontali (`plt.barh`) ---
plt.barh(x, y1, label='italiani', color='red')

plt.title('nome del grafico') 
plt.xlabel('eta')
plt.ylabel('valori caso')
plt.legend()
plt.show()
```
