```python
# Importa la libreria seaborn per visualizzazioni statistiche.
import seaborn as sns 

# Importa la libreria pandas per la manipolazione e l'analisi dei dati.
import pandas as pd 

# Importa la libreria matplotlib.pyplot per la creazione di grafici.
import matplotlib.pyplot as plt 
```

```python
# Carica il dataset 'tips' 
# (mance) integrato in seaborn in un DataFrame di pandas
df = sns.load_dataset("tips")

# Stampa le prime 5 righe del DataFrame per dare un'occhiata ai dati.
print(df.head())
```

```python
# Crea un box plot (grafico a scatola e baffi).
# 'x="day"' specifica la variabile da usare 
# sull'asse x (i giorni della settimana).
# 'y="total_bill"' specifica la variabile da 
# usare sull'asse y (l'importo totale del conto).
# 'data=df' specifica il DataFrame da cui prendere i dati.
sns.boxplot(x="day", y="total_bill", data=df)
plt.title("Box Plot: Total Bill per Giorno")
plt.show()
```
