```python
import matplotlib.pyplot as plt
```

```python
# --- Dati di Base ---
slices = [60,40,50,30,10]  # Dimensioni relative delle fette (somma 190).
labels = ['si','no','forse','non so','non so dire'] # Etichette per ogni fetta.
colors = ['red','brown', 'orange','green','blue'] # Colori assegnati a ciascuna fetta.
explode = [0.2,0.1,0.3,0.4,0.5] # Distanza di ogni fetta dal centro (0.0 non è staccata, 0.5 è molto staccata).

# Grafico 1: Torta con Fette Esplose (Exploded Pie Chart)
plt.pie(slices, # La lista dei valori che definiscono la dimensione delle fette.
        labels=labels, # Assegna le etichette alle fette.
        colors=colors, # Assegna i colori definiti.
        wedgeprops={'edgecolor': '#000000'}, # Proprietà degli "spicchi": imposta un bordo nero (esadecimale) attorno a ogni fetta.
        startangle=70, # Angolo da cui inizia il disegno della prima fetta (di default è 0, cioè l'asse X positivo).
        explode=explode) # Applica l'effetto di "esplosione" (stacco dal centro) definito dalla lista.

plt.title('nome del grafico')
plt.show()
```

```python
slices = [60,40,50,30,10]
labels = ['si','no','forse','non so','non so dire']
colors = ['red','brown', 'orange','green','blue']
explode = [0.2,0.1,0.3,0.4,0.5]
plt.pie(slices, 
        labels=labels, 
        colors=colors, 
        wedgeprops={'edgecolor': '#000000'}, 
        startangle=70, 
        autopct='%1.1f%%' )# Calcola e visualizza le percentuali. Il formato '%1.1f%%' significa:
                          # '%': segno di inizio format string.
                          # '1.1': un numero intero (prima del punto) e un decimale (dopo il punto).
                          # 'f': float (numero decimale).
                          # '%%': aggiunge il simbolo di percentuale alla fine.

plt.title('nome del grafico')
plt.show()
```
