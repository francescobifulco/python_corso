```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [3,6,7,2,5]
```

```python
# Sezione 1: Interfaccia Basata sullo Stato vs. Oggetti

plt.plot(x,y)
plt.title('nome del grafico')
plt.xlabel('Prova')
```

```python
# Interfaccia Orientata agli Oggetti (Consigliata)

fig, ax = plt.subplots() # Crea esplicitamente una Figura (fig) e un Asse (ax) di plotting.
ax.plot(x,y) # Chiama il metodo plot sull'oggetto Asse (ax).
ax.set_title('Grafico Singolo (Oggetto Ax)') # Imposta il titolo usando il metodo set_title().
ax.set_xlabel('Prova')  # Imposta l'etichetta il metodo set_xlabel().
plt.show()
```

```python
# Sezione 2: Creazione di Sub-Grafici (Row-Column Layout)

# Creazione di 2 sub-grafici in una singola colonna (2 righe, 1 colonna).
# L'unpacking assegna ogni asse (ax) a una variabile separata (ax1 e ax2).
fig, ax = plt.subplots(nrows=2, ncols=1)
```

```python
# Creazione di 2 sub-grafici in una singola colonna (2 righe, 1 colonna).
# L'unpacking assegna ogni asse (ax) a una variabile separata (ax1 e ax2).

fig, (ax1, ax2 ) = plt.subplots(nrows=2, ncols=1, sharex=True)
ax1.plot(x,y)
ax1.set_title('nome del grafico')
ax1.set_xlabel('Prova')

ax2.plot(x,y)

ax2.set_xlabel('Prova')
```

```python
fig, ax1 = plt.subplots()
fig, ax2 = plt.subplots()
ax1.plot(x,y)
ax1.set_title('nome del grafico')
ax1.set_xlabel('Prova')

ax2.plot(x,y)
ax2.set_title('nome del grafico')
ax2.set_xlabel('Prova')
```

```python
fig, (ax1a, ax1b) = plt.subplots(nrows=2, ncols=1)
fig, ax2 = plt.subplots()
ax1a.plot(x,y)
ax1a.set_title('nome del grafico')
ax1a.set_xlabel('Prova')

ax2.plot(x,y)
ax2.set_title('nome del grafico')
ax2.set_xlabel('Prova')
```
