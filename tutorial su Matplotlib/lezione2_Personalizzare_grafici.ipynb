```python
import matplotlib.pyplot as plt
```

```python
# --- Dati per il grafico ---
x = [25, 26, 27, 28, 29, 30, 31]
y1 = [100, 120, 130, 140, 150, 145, 140]
y2 = [90,80,100,120,130,200,300]
```

```python
# --- Primo Grafico: Personalizzazione Standard (Senza Stile) ---

#plt.plot(x, y1, label='italiano', color='red', linewidth=5, linestyle='--')

plt.plot(x, y1, # Dati per l'asse X e Y.
         label='italiani',  # Etichetta da usare nella Legenda (corretto da 'italiano').
         color='red',  # Colore della linea (usando il nome del colore).
         marker='o')  # Forma del marcatore per ogni punto ('o' è un cerchio).

plt.plot(x, y2,                # Dati per l'asse X e Y.
         label='francesi',     # Etichetta da usare nella Legenda.
         color='#212121',    # Colore della linea (usando il codice esadecimale per il nero quasi assoluto).
         linestyle='dotted')   # Stile della linea (linea tratteggiata con puntini).

plt.title('nome del grafico') # Imposta il titolo principale del grafico.
plt.xlabel('eta') # Etichetta per l'asse delle ascisse (X).
plt.ylabel('valori caso')  # Etichetta per l'asse delle ordinate (Y).
plt.legend(['italiani', 'francesi'])
plt.grid() # Aggiunge una griglia di sfondo al grafico (impostato su True per attivarla).
plt.show()
```

```python
# --- Esplorazione degli Stili ---

# Stampa tutti gli stili disponibili nella libreria.
print(plt.style.available)
```

```python
# Applica lo stile 'Solarize_Light2' (chiaro e contrastato).

plt.style.use('Solarize_Light2')
plt.plot(x, y1, label='italiani', color='red', marker='o')
plt.plot(x,y2, label='francesi', color='#212121', linestyle='dotted')
plt.show()
```

```python
# Applica lo stile 'seaborn-v0_8-colorblind' (ottimizzato per daltonici).
 
plt.style.use('seaborn-v0_8-colorblind')
plt.plot(x, y1, label='italiani', color='red', marker='o')
plt.plot(x,y2, label='francesi', color='#212121', linestyle='dotted')
plt.show()
```

```python
# Applica lo stile 'ggplot' (ispirato al popolare pacchetto di R).

plt.style.use('ggplot')
plt.plot(x, y1, label='italiani', color='red', marker='o')
plt.plot(x,y2, label='francesi', color='#212121', linestyle='dotted')
plt.show()
```
