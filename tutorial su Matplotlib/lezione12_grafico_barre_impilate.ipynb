```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Definisce le etichette per l'asse categorico (i trimestri)
trimestri = ['Q1', 'Q2', 'Q3', 'Q4']

# Ricavi dei due prodotti per ogni trimestre
ricavi_prodottoA = [25000, 30000, 28000, 35000]
ricavi_prodottoB = [18000, 20000, 22500, 25000]
```

```python
# 1. GRAFICO A BARRE ACCATASTATE VERTICALE (Stacked Bar Chart)

# Barra per il Prodotto A (la base)
plt.bar(trimestri, ricavi_prodottoA, label='Prodotto A')

# Barra per il Prodotto B (accatastata sopra il Prodotto A)
# L'argomento 'bottom=ricavi_prodottoA' indica che la barra B inizia dove finisce la barra A
plt.bar(trimestri, ricavi_prodottoB, bottom=ricavi_prodottoA, label='Prodotto B')
plt.xlabel('Trimestre')
plt.ylabel('Ricavi (€)')
plt.title('Ricavi trimestrali di due prodotti')

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7) # Aggiunge una griglia sull'asse y

# Aggiunge le etichette dei valori su ogni segmento e il totale in cima
for i in range(len(trimestri)):
    totale = ricavi_prodottoA[i] + ricavi_prodottoB[i]
    
    # Etichetta per Prodotto A (posizionata al centro della sua barra)
    plt.text(i, ricavi_prodottoA[i] / 2, f"{ricavi_prodottoA[i]:,}", ha='center', va='center', color='white')
    
    # Etichetta per Prodotto B (posizionata al centro della sua barra, sopra la barra A)
    plt.text(i, ricavi_prodottoA[i] + ricavi_prodottoB[i] / 2, f"{ricavi_prodottoB[i]:,}", ha='center', va='center', color='white')
    
    # Etichetta per il TOTALE (posizionata leggermente sopra la barra B)
    plt.text(i, totale + 1000, f"Tot: {totale:,}", ha='center', va='bottom', fontsize=8, color='black')
    
plt.tight_layout() # Adatta automaticamente i parametri della figura
plt.show()
```

```python
# 2. GRAFICO A BARRE ACCATASTATE ORIZZONTALE (Horizontal Stacked Bar Chart)
# (Corretto con etichette asse X/Y appropriate)

# Barra per il Prodotto A (la base)
# plt.barh usa l'argomento 'left' per l'accatastamento orizzontale
plt.barh(trimestri, ricavi_prodottoA, label='Prodotto A')
plt.barh(trimestri, ricavi_prodottoB, left=ricavi_prodottoA, label='Prodotto B')
plt.xlabel('Trimestre')
plt.ylabel('Ricavi (€)')
plt.title('Ricavi trimestrali di due prodotti')

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
