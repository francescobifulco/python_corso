Obiettivi:
Creare una classe base ArticoloBiblioteca con attributi protetti (_titolo, _autore, _anno_pubblicazione) e metodi pubblici per accedere e modificare questi attributi.

Creare una classe derivata Libro che erediti da ArticoloBiblioteca e aggiunga un attributo privato __isbn.

Creare un'altra classe derivata Rivista che erediti da ArticoloBiblioteca e aggiunga un attributo privato __numero_edizione.

Implementare il polimorfismo creando una funzione descrivi_articolo(articolo) che accetti un oggetto di tipo ArticoloBiblioteca (o una delle sue sottoclassi) e chiami un metodo polimorfico ottieni_dettagli() che sarà definito in ogni classe per restituire i dettagli specifici dell'articolo.

Implementare l'overriding dei metodi assicurandosi che ottieni_dettagli() in Libro e Rivista non solo restituisca i dettagli dell'articolo generico, ma anche quelli specifici della sottoclasse.

Dimostrare l'incapsulamento provando ad accedere direttamente agli attributi protetti/privati dall'esterno della classe (e osservando gli errori o le convenzioni).

```python
class ArticoloBiblioteca:
    def __init__(self, titolo, autore, anno_pubblicazione):
        self._titolo = titolo
        self._autore = autore
        self._anno_pubblicazione = anno_pubblicazione
    def get_titolo(self):
        return self._titolo

    def get_autore(self):
        return self._autore

    def get_anno_pubblicazione(self):
        return self._anno_pubblicazione
    
    def ottieni_dettagli(self):
        return (f"Titolo: {self._titolo}, Autore: {self._autore}, "
                f"Anno: {self._anno_pubblicazione}")
```

```python
class Libro(ArticoloBiblioteca):
    def __init__(self, titolo, autore, anno_pubblicazione, isbn):
        super().__init__(titolo, autore, anno_pubblicazione)
        self.__isbn = isbn
    def get_isbn(self):
        return self.__isbn
    def ottieni_dettagli(self):
        return f"{super().ottieni_dettagli()}, ISBN: {self.__isbn}"
```

```python
class Rivista(ArticoloBiblioteca):
    def __init__(self, titolo, autore, anno_pubblicazione, numero_edizione):
        super().__init__(titolo, autore, anno_pubblicazione)
        self.__numero_edizione = numero_edizione
    def ottieni_dettagli(self):
        return f"{super().ottieni_dettagli()}, Numero Edizione: {self.__numero_edizione}"
```
