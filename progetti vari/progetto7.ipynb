Stringhe con join():

Data una lista di interessi (ad esempio, ['programmazione', 'lettura', 'escursioni']), unirli in una singola stringa, separandoli con una virgola e uno spazio (es. "programmazione, lettura, escursioni").

```python
dizionario = {'nome': ["Luca Rossi", "Giulia Bianchi", "Marco Verdi", "Sofia Gialli", "Andrea Neri", "Francesca Romano", "Alessandro Ferrari", "Chiara Marino", "Davide Esposito", "Sara Ricci"],
              'eta':  [25, 32, 41, 28, 35, 22, 48, 38, 29, 42],
              'citta': ["Roma", "Milano", "Napoli", "Torino", "Firenze", "Venezia", "Bologna", "Palermo", "Genova", "Bari"],
              'interessi': ["Musica", "Sport", "Lettura", "Viaggi", "Cucina", "Fotografia", "Tecnologia", "Arte", "Teatro", "Danza"]
}
dizionario
```

```python
import json

with open("dati.json", "w") as file:
    json.dump(dizionario, file, indent=4)
```

```python
with open('dati.json', 'r') as file:
    contenuti = file.read()
contenuti
```

```python
with open('dati.json', 'r') as file:
    dizionario = json.load(file)
dizionario
```

```python
lista_di_interessi_da_unire = ["programmazione", "lettura", "escursioni", "cucina asiatica"]
print(f"Lista originale di interessi: {lista_di_interessi_da_unire}")

# Unire la lista in una singola stringa con ", " come separatore
stringa_interessi = ", ".join(lista_di_interessi_da_unire)

print(f"Stringa di interessi unita: \"{stringa_interessi}\"")
```
