```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta
```

```python
tempo = np.random.randint(0,50, size=15)
tempo
```

```python
# Data di inizio e fine
data_inizio = datetime(2025, 7, 1)
data_fine = datetime(2025, 7, 17)

# Generare la sequenza
sequenza_date = [data_inizio + timedelta(days=i) for i in range((data_fine - data_inizio).days + 1)]

# Stampare le date
for data in sequenza_date:
    print(data.strftime("%Y-%m-%d"))
```

```python
data = ['2025-07-01','2025-07-02','2025-07-03','2025-07-04',
        '2025-07-05','2025-07-06','2025-07-07','2025-07-08',
        '2025-07-09','2025-07-10','2025-07-11','2025-07-12',
        '2025-07-13','2025-07-14','2025-07-15']
data
```

```python
df = pd.DataFrame()
df['temperature'] = tempo
df
```

```python
df['date'] = data
df 
```

```python
df
```

```python
def classifica_temperatura(temp):
    if temp > 25:
        return "Calda"
    elif 15 <= temp <= 25:
        return "Moderata"
    else: # temp < 15
        return "Fredda"
df['livello di calore'] = df['temperature'].apply(classifica_temperatura)
df
```

```python
media = df['temperature'].mean()
print(f"Temperatura Media: {media:.2f}°C")
```

```python
massima = df['temperature'].max()
print(f"Temperatura Massima: {massima}°C")
```

```python
minima = df['temperature'].min()
print(f"Temperatura Minima: {minima}°C")
```
