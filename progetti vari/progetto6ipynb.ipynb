```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt
```

```python
vendite_mensili = np.random.randint(0,100, size=55)
vendite_mensili
```

```python
df = pd.DataFrame()
df['vendite mensile'] = vendite_mensili
df
```

```python
df['categorie'] = categorie
df
```

```python
try:
    df.to_csv("vendite.csv", index=False)
    df = pd.read_csv('vendite.csv')
except FileNotFoundError:
    print('file non esiste')
df
```

```python
# Creazione del grafico
plt.figure(figsize=(10, 6))
plt.plot(df.index, df['vendite mensile'], color='blue', linestyle='--', marker='o', label='Linea 1')
plt.title("Andamento delle Vendite nel Tempo")
plt.xlabel("Mese")
plt.ylabel("Vendite")
plt.legend()
plt.grid(True)

# Mostra il grafico
plt.show()
```

```python
#Raggruppiamo per 'categorie' e calcoliamo la somma delle 
#vendite per ogni categoria
vendite_per_categoria = df.groupby('categorie')['vendite mensile'].sum().reset_index()
```

```python
import seaborn as sns
plt.figure(figsize=(12, 7)) # Dimensione del grafico più grande per categorie
sns.barplot(x='categorie', y='vendite mensile', data=vendite_per_categoria, palette='viridis')
plt.title("Vendite Totali per Categoria di Prodotto")
plt.xlabel("Categoria di Prodotto")
plt.ylabel("Totale Vendite")
plt.xticks(rotation=45, ha='right') # Ruota le etichette dell'asse X per leggibilità
plt.tight_layout() # Ottimizza il layout per prevenire sovrapposizioni
plt.show()
```
