Progetto 3: Gestione Inventario Semplice (Funzioni, Classi e Oggetti) 📦
Questo progetto introduce i concetti fondamentali della programmazione orientata agli oggetti (OOP) con funzioni, classi, attributi e metodi.

Obiettivi:
Definire una classe Prodotto con gli attributi: nome, prezzo e quantita.

Aggiungere un metodo get_info() alla classe Prodotto che ritorni una stringa con tutti i dettagli del prodotto.

Aggiungere un metodo vendi(quantita_venduta) che riduca la quantita disponibile e restituisca True se la vendita è possibile (c'è abbastanza stock), False altrimenti.

Aggiungere un metodo rifornisci(quantita_aggiunta) che aumenti la quantita disponibile.

Creare alcuni oggetti della classe Prodotto.

Scrivere una funzione mostra_inventario(lista_prodotti) che prenda una lista di oggetti Prodotto e stampi le informazioni di ogni prodotto usando il metodo get_info().

Simulare delle operazioni di vendita e rifornimento, chiamando i metodi appropriati.

```python
class Prodotto:
    def __init__(self,nome,prezzo,quantita):
        self.nome = nome
        self.prezzo = prezzo
        self.quantita = quantita
    def get_info(self):     
        return f'il prodotto {self.nome} costa {self.prezzo} ci sono {self.quantita}'
    def vendi(self,quantita_venduta):
        if self.quantita >= quantita_venduta:
            print('ci sono abbastanza di prodotti')
            self.quantita -= quantita_venduta
            print(f"Venduti {quantita_venduta} unità di {self.nome}. Nuova quantità disponibile: {self.quantita}")
            return True
        else:
            print(f"Non ci sono abbastanza {self.nome} in stock. Disponibili: {self.quantita}, Richiesti: {quantita_venduta}")
            return False
    def rifornisci(self,quantita_aggiunta):
            self.quantita += quantita_aggiunta
            print(f"Aggiunte {quantita_aggiunta} unità a {self.nome}. Nuova quantità disponibile: {self.quantita}")
```

```python
def mostra_inventario(lista_prodotti):
    for prodotto in lista_prodotti:
        print(prodotto.get_info())
```

```python
p1 = Prodotto('pc',1500,5000)
p1.get_info()
p1.vendi(1000)
p1.rifornisci(500)
prodotti = [
    Prodotto("Telefono", 500, 10),
    Prodotto("Computer", 1000, 5),
    Prodotto("Tablet", 300, 20),
    p1
]
mostra_inventario(prodotti)
```
