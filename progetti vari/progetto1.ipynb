Progetto 1: Calcolatrice Semplice e Gestione Dati Personali 💻
Questo progetto introduce i concetti base di input/output, operazioni aritmetiche, variabili, tipi di dati e stringhe.

Obiettivi:
Chiedere all'utente il suo nome e la sua età.

Stampare un messaggio di benvenuto personalizzato.

Chiedere due numeri all'utente.

Calcolare la somma, la differenza, il prodotto e il quoziente di questi due numeri.

Stampare i risultati delle operazioni in un formato leggibile.

Utilizzare variabili per memorizzare i dati e i risultati.

Assicurarsi che i tipi di dati siano gestiti correttamente (ad esempio, convertire l'input numerico).

Utilizzare alcuni metodi delle stringhe per formattare l'output (es. .upper(), .capitalize()).

```python
nome = str(input('inserire il tuo nome: ')).capitalize()
eta = int(input('inserisci la tua eta: '))
print(f'ti do il benvenuto {nome} e la tua eta e {eta}')
```

```python
num1 = int(input('inserisci un numero: '))
num2 = int(input('inserisci un numero: '))

somma = num1 + num2
differenza = num1 - num2
prodotto = num1 * num2
quoziente = num1 // num2

print(f'la somma tra {num1} e {num2}: {somma}')
print(f'la differenza tra {num1} e {num2}: {differenza}')
print(f'il prodotto tra {num1} e {num2}: {prodotto}')
print(f'il quoziente tra {num1} e {num2}: {quoziente}')
```
