Progetto 2: Analisi Voti Studenti (Liste, Dizionari, Condizioni e Cicli) 📝
Questo progetto introduce l'uso di strutture dati complesse come liste e dizionari, insieme a costrutti di controllo del flusso come if/else/elif e cicli for/while.

Obiettivi:
Creare una lista di nomi di studenti.

Creare un dizionario dove le chiavi sono i nomi degli studenti e i valori sono liste dei loro voti (ad esempio, {'Alice': [8, 7, 9], 'Bob': [6, 8, 7]}).

Utilizzare un ciclo for per iterare su ogni studente nel dizionario.

Per ogni studente, calcolare la media dei suoi voti.

Utilizzare if/else/elif per determinare il giudizio dello studente in base alla media:

Media >= 9: "Eccellente"

Media >= 7: "Buono"

Media >= 6: "Sufficiente"

Media < 6: "Non Sufficiente"

Stampare il nome dello studente, la sua media e il suo giudizio.

Aggiungere la possibilità di uscire dal programma con la parola chiave break in un ciclo while che chiede di continuare ad analizzare nuovi studenti.

Gestire il caso in cui uno studente non abbia voti (usando continue o pass in un blocco condizionale).

```python
studenti = [{"nome": "Luca", "cognome": "Rossi", "età": 18, "classe": "4A"},
            {"nome": "Giulia", "cognome": "Bianchi", "età": 17, "classe": "3B"},
            {"nome": "Marco", "cognome": "Verdi", "età": 19, "classe": "5C"},
            {"nome": "Sofia", "cognome": "Gialli", "età": 16, "classe": "2A"},
            {"nome": "Andrea", "cognome": "Neri", "età": 18, "classe": "4B"}
]
studenti
```

```python
dizionario = {"Luca Rossi": [8, 7, 9],
              "Giulia Bianchi": [5, 4, 7],
              "Marco Verdi": [10, 3, 6],
              "Sofia Gialli": [7, 7, 7],
              "Andrea Neri": [10, 10, 10],
              "Lucia Esposito" : [0, 0, 0]
              }
dizionario
```

```python
for chiave,valore in dizionario.items():
    print(f'{chiave}: {valore}')
```

```python
from statistics import mean
for valore in dizionario.values():
    media = mean(valore)
    print(f'la media dei voto per ogni studente: {media:.2f}')
```

```python
print(f'La lista degli studenti: {list(dizionario.keys())}')
while True:    
    scelta = input('Inserire il nome dello studente o "esci" per uscire: ')
    if scelta.lower() == 'esci': break
    elif scelta in dizionario.keys():
         voti = dizionario[scelta]
         media = sum(voti) / len(voti)
         if media >= 9: print(f'{scelta} Eccellente la media dei voti e {media}')
         elif media >= 7: print(f'{scelta} Buono la media dei voti e {media}')
         elif media >= 6: print(f'{scelta} Sufficiente la media dei voti e {media}')
         elif media >= 1 and media < 6: print(f'{scelta} Non Sufficiente la media dei voti e {media}')
         elif media == 0.0: 
             print(f'{scelta} non ha voti')
             continue
```
