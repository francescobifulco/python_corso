## Livello Base: Introduzione e Visualizzazioni Univariate e Bivariate Semplici
L'obiettivo è comprendere come importare la libreria, caricare i dati e creare le visualizzazioni più comuni.

Setup Iniziale: Installa Seaborn (pip install seaborn) e importalo con la convenzione standard (import seaborn as sns). Importa anche Pandas e Matplotlib (import pandas as pd, import matplotlib.pyplot as plt).

Caricamento Dati: Utilizza uno dei dataset integrati di Seaborn (es. tips, iris, titanic) per familiarizzare con l'accesso ai dati.

Esempio: df = sns.load_dataset("tips")

Distribuzione di una Singola Variabile (Univariate):

Crea un istogramma per visualizzare la distribuzione di una variabile numerica (sns.histplot).

Utilizza un KDE plot (sns.kdeplot) per stimare la funzione di densità di probabilità.

Relazione tra Due Variabili (Bivariate):

Crea un diagramma a dispersione (scatterplot) per vedere la relazione tra due variabili numeriche (sns.scatterplot o sns.relplot).

Aggiungi una retta di regressione (sns.regplot) per stimare la relazione lineare.

Visualizzazione Dati Categorici:

Usa un bar plot (sns.barplot) per visualizzare statistiche (es. media) per diverse categorie.

Crea un box plot (sns.boxplot) per visualizzare la distribuzione, i quartili e gli outlier di una variabile numerica per diverse categorie.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("titanic")
df.head()
```

```python
sns.histplot(data=df, x="fare", kde=True)
```

```python
sns.kdeplot(data=df, x="fare", fill=True)
```

```python
sns.scatterplot(data=df, x="fare", y="age", hue="sex")
```

```python
sns.regplot(data=df, x="fare", y="age")
```

```python
sns.barplot(data=df, x="pclass", y="age", estimator=sum)
```

```python
sns.boxplot(data=df, x="age", y="fare")
plt.title("Distribuzione della tariffa per classe")
plt.xlabel("Classe")
plt.ylabel("Tariffa ($)")
plt.ylim(0, 600)
```
