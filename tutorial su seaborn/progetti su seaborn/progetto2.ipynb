## Livello Intermedio: Analisi Multivariata e Personalizzazione

L'obiettivo è estendere le visualizzazioni per includere più variabili e affinare l'estetica dei grafici.

Integrazione di Variabili Aggiuntive:

Utilizza i parametri hue, size, e style nelle funzioni come sns.scatterplot o sns.relplot per includere una terza e una quarta variabile (categorica o numerica).

Grafici per Distribuzioni Dati Categorici Avanzati:

Sperimenta con i grafici a violino (sns.violinplot) che combinano i box plot con una stima della densità kernel.

Crea strip plot o swarm plot (sns.stripplot, sns.swarmplot) per visualizzare tutti i punti dati, utili per dataset più piccoli.

Matrici di Correlazione e Heatmap:

Calcola la matrice di correlazione del DataFrame (df.corr()).

Visualizza la matrice di correlazione utilizzando una heatmap (sns.heatmap) per identificare rapidamente le relazioni tra tutte le coppie di variabili numeriche.

Personalizzazione Estetica:

Applica i temi di Seaborn (sns.set_theme(), es. 'whitegrid').

Utilizza diverse palette di colori con il parametro palette.

Personalizza i titoli e le etichette degli assi usando funzioni di Matplotlib come plt.title(), plt.xlabel(), e plt.ylabel().

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("iris")
df.head()
```

```python
sns.relplot(data=df, x='species', y='petal_length', hue='sepal_length', size=5, style='sepal_length')
```

```python
sns.violinplot(data=df, x='species', y='petal_length')
```

```python
sns.swarmplot(data=df, x='species', y='petal_width')
```

```python
correlation_matrix = df.corr(numeric_only=True)
correlation_matrix
```

```python
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5, vmin=-1, vmax=1)
plt.title("Matrice di Correlazione")
plt.show()
```

```python
sns.set_theme(style="whitegrid")
sns.boxplot(data=df, x="species", y="petal_length")
plt.title("Distribuzione della lunghezza dei petali")
plt.show()
```
