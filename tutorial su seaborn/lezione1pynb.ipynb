```python
import seaborn as sns  # Libreria per grafici statistici
import pandas as pd
```

```python
# Carica un dataset campione con dati di mance nei ristoranti
df = sns.load_dataset('tips')
df.head()
```

```python
# Istogramma con curva KDE (densità)
sns.histplot(data=df, x="total_bill", kde=True)
```

```python
# Solo curva KDE riempita
sns.kdeplot(data=df, x="total_bill", fill=True)
```

```python
sns.boxplot(data=df, x="day", y="total_bill") # Boxplot per giorno
```

```python
# Violin plot per giorno, diviso per sesso
sns.violinplot(data=df, x="day", y="total_bill", hue="sex", split=True)
```

```python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex")
# Scatter plot con colori diversi per sesso
```

```python
sns.lineplot(data=df, x="size", y="tip", ci=None)
# Line plot (size vs tip), senza intervallo di confidenza
```

```python
# Scatter plot multiplo (colonne per time: Lunch/Dinner), colori per sesso
sns.relplot(data=df, x="total_bill", y="tip", hue="sex", col="time", kind="scatter")
```

```python
sns.barplot(data=df, x="day", y="tip", estimator=sum)
# Barre per giorno, altezza basata sulla somma delle mance
```

```python
sns.countplot(data=df, x="day", hue="sex")
# Conteggio delle osservazioni per giorno e sesso
```

```python
sns.stripplot(data=df, x="day", y="total_bill", jitter=True)
# Punti individuali con "jitter" per evitare sovrapposizione
```

```python
sns.swarmplot(data=df, x="day", y="total_bill", hue="sex")
# Come stripplot, ma evita la sovrapposizione in automatico
```

```python
sns.regplot(data=df, x="total_bill", y="tip")
# Regressione lineare con scatterplot e linea di regressione
```

```python
sns.lmplot(data=df, x="total_bill", y="tip", hue="sex", col="time")
# Versione ad alto livello di regplot, separata per time e sesso
```

```python
sns.pairplot(sns.load_dataset("iris"), hue="species")
# Grafico a matrice per ogni coppia di variabili, colori per specie
```

```python
sns.boxplot(data=df, x="day", y="total_bill")
plt.title("Distribuzione conto per giorno")
plt.xlabel("Giorno della settimana")
plt.ylabel("Totale conto ($)")
plt.ylim(0, 60)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="total_bill", bins=20)
plt.savefig("grafico.png") # Salva l’immagine su file
```

```python
# Griglia con colonne per sesso e righe per orario
g = sns.FacetGrid(df, col="sex", row="time")

# Applica scatterplot in ogni pannello
g.map(sns.scatterplot, "total_bill", "tip")
```

```python
sns.boxplot(data=df, x="day", y="total_bill", whis=[0, 100])
# Boxplot che mostra tutti i valori (whiskers 0–100%)

sns.swarmplot(data=df, x="day", y="total_bill", color=".25")
# Punti individuali sovrapposti in trasparenza per aggiungere dettaglio
```
