Crea uno scatter plot in cui ogni punto ha un marcatore diverso, dimensioni variabili e bordo colorato.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
asseX = np.arange(10)
asseY = np.random.randint(0, 11, size=10)
```

```python
colori = np.random.choice(['red', 'green', 'blue', 'orange', 'purple'], size=10)
marcatori = ['x', 'o', '^', '*', '+', 's', 'D', 'P', '<', '>']
dimensione = np.random.randint(50, 80, size=10)
```

```python
for i in range(len(asseX)):
    plt.scatter(
        asseX[i],
        asseY[i],
        color=colori[i],
        marker=marcatori[i],
        s=dimensione[i],
        edgecolor='black',
        linewidth=1.2
    )
plt.title('Personalizzazione marcatori')
plt.xlabel('Valori')
plt.ylabel('Punti')
plt.grid()
plt.show()
```
