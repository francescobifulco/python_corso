Disegna la funzione y = x² per x da −10 a 10, usando una linea tratteggiata, spessore personalizzato e colore personalizzato.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-10, 10, 400)
y = x ** 2
```

```python
plt.plot(y, label='Valori di Y in funzione di X', 
         color='red', linestyle='--', linewidth=2.5)
plt.title('Modifica dello stile della linea')
plt.xlabel('Asse X')
plt.ylabel('Asse Y')
plt.grid()
plt.legend()
plt.show()
```
