## Esercizio 5
Genera 100 punti casuali usando `numpy.random.randn` e plottali come scatter con colore in alpha blending.

```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
punti = np.random.randn(100)
x = np.arange(100)
```

```python
plt.scatter(x, punti, c='red', alpha=0.5, label='Punti casuali')
plt.title('Grafico scatter di base')
plt.xlabel('Valori')
plt.ylabel('Punti')
plt.grid()
plt.legend()
plt.show()
```
