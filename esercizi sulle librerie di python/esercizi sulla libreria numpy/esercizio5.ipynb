## Esercizio 5
Crea due array casuali 4×4 (valori interi tra 0 e 20).
Calcola la somma, differenza, prodotto elemento per elemento e divisione.
Trova gli indici dei valori maggiori nella differenza.

```python
import numpy as np
```

```python
v1 = np.random.randint(0, 21, (4,4))
v2 = np.random.randint(0, 21, (4,4))
```

```python
somma = np.add(v1, v2)
somma
```

```python
differenza = np.subtract(v1, v2)
differenza
```

```python
indici_max = np.argmax(differenza)
indici_max
```

```python
prodotto = np.multiply(v1, v2)
prodotto
```

```python
divisione = np.divide(v1,v2) 
divisione
```
