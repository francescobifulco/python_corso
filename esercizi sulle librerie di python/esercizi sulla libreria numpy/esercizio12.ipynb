## Esercizio 12
* Crea due matrici A (3×4) e B (4×3) con numeri interi casuali.
* Esegui la moltiplicazione matriciale `A @ B`.
* Calcola la traccia e la somma degli elementi sulla diagonale.

```python
import numpy as np 
```

```python
A = np.random.randint(0, 11, (3,4))
B = np.random.randint(11, 21, (4,3))
```

```python
C = A @ B
C
```

```python
diagonale = np.diag(C)
diagonale
```

```python
traccia = np.trace(C)
traccia
```

```python
somma_diagonale = np.sum(diagonale)
somma_diagonale
```
