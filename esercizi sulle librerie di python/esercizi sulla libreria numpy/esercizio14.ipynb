## Esercizio 14
* Crea una matrice 5×5 casuale.
* Trasponila e confronta se è simmetrica.
* Rendi la matrice simmetrica forzatamente e calcola la media dei valori sopra la diagonale.

```python
import numpy as np
```

```python
v1 = np.random.randint(0, 11, (5,5))
v1
```

```python
v_T = v1.T
v_T
```

```python
verifica = np.allclose(v1, v_T)
verifica
```

```python
v_sym = (v1 + v_T) / 2
v_sym
```

```python
media_dia = v_sym[np.triu_indices(5, k=1)]
media = np.mean(media_dia)
media
```
