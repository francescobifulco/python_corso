Crea un array 6×6 di zeri e riempi la diagonale principale e secondaria con 1.
Somma tutti gli elementi della matrice e stampa il risultato.

```python
import numpy as np
```

```python
diagonale_prin = np.eye(6)
diagonale_prin
```

```python
diagonale_seco = np.fliplr(np.eye(6))
diagonale_seco
```

```python
tot_diago = np.array(diagonale_prin + diagonale_seco)
tot_diago
```

```python
somma = np.sum(tot_diago)
print("\nSomma di tutti gli elementi:", somma)
```
