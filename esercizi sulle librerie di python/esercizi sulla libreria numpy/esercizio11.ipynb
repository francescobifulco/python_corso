## Esercizio 11
* Crea un array casuale di 20 numeri float.
* Ordina l’array in ordine decrescente.
* Trova la posizione del 3° valore più grande e sostituiscilo con la media di tutti i valori.

```python
import numpy as np
```

```python
v1 = np.random.random(size=20)
v1
```

```python
ordina = np.sort(v1)[::-1]
ordina
```

```python
terzo_valore = ordina[2]
terzo_valore
```

```python
posizione = np.where(v1 == terzo_valore)[0][0]
posizione
```

```python
media = np.mean(v1)
v1[posizione] = media
v1
```
