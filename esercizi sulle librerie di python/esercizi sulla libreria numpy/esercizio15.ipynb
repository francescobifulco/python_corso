## Esercizio 15
* Crea una matrice 10×10 di zeri.
* Inserisci casualmente 10 valori diversi da zero.
* Conta quanti valori non nulli ci sono e sostituiscili con la loro radice quadrata.


```python
import numpy as np
```

```python
v1 = np.zeros((10,10))
v1
```

```python
valori_casuali = np.random.randint(1, 101, size=10)
valori_casuali
```

```python
righe = np.random.randint(0, 10, size=10)
colonne = np.random.randint(0, 10, size=10)
```

```python
for i in range(10):
    v1[righe[i], colonne[i]] = valori_casuali[i]

print("\nMatrice con valori casuali inseriti:\n", v1)

non_zeri = np.count_nonzero(v1)
print("\nNumero di valori non nulli:", non_zeri)

v1[v1 != 0] = np.sqrt(v1[v1 != 0])
print("\nMatrice dopo la radice quadrata dei valori non nulli:\n", v1)
```
