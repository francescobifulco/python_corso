* Genera una matrice 8×8 con numeri casuali tra 0 e 1.
* Calcola: media globale, somma per riga, valore minimo e massimo per colonna.
* Crea un array con la posizione dell’elemento massimo globale.


```python
import numpy as np
```

```python
v1 = np.random.uniform(0, 1, (8,8))
v1
```

```python
media = np.mean(v1)
media
```

```python
somma_riga = np.sum(v1, axis=1)
somma_riga
```

```python
valore_mass = np.max(v1, axis=0)
valore_mass
```

```python
valore_mini= np.min(v1, axis=0)
valore_mini
```

```python
indice_massimo_globale = np.unravel_index(np.argmax(v1), v1.shape)
indice_massimo_globale
```
