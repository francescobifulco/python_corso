## Esercizio 13
* Genera una matrice quadrata 4×4 casuale.
* Calcola il determinante e verifica se la matrice è invertibile.
* Se sì, calcola la matrice inversa e verifica la correttezza tramite `A @ A_inv ≈ I`.


```python
import numpy as np
```

```python
v1 = np.random.randint(0, 11, (4,4))
v1
```

```python
determinante = np.linalg.det(v1)
determinante
```

```python
if determinante != 0:
    print("La matrice è invertibile.")
    
    inv = np.linalg.inv(v1)
    print(f"La matrice inversa: {inv}")
    
    controllo = v1 @ inv
    print(f"controlla del inversa: \n{controllo}")
    
    print(f"LA verivica: {np.allclose(controllo, np.eye(4))}")
    
else:
    print("La matrice NON è invertibile.")
```
