## Esercizio 18
* Genera una matrice quadrata 3×3 casuale.
* Calcola gli autovalori e autovettori.
* Verifica la relazione `A @ v = λv` per ciascuna coppia.

```python
import numpy as np
```

```python
v1 = np.random.uniform(0, 11, (3,3))
v1
```

```python
autovalori, autovettori = np.linalg.eig(v1)
autovalori, autovettori
```

```python
print("\nVerifica v1 @ v = λ v:\n")
for i in range(len(autovalori)):
    λ = autovalori[i]
    v = autovettori[:, i]
    
    left = v1 @ v                  
    right = λ * v              

    print(f"Autovalore λ{i+1} = {λ}")
    print("v1 @ v  =", left)
    print("λ * v  =", right)
    print("Differenza = ", left - right)
    print("-" * 40)
```
