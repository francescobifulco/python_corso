* Crea due vettori tridimensionali casuali.
* Calcola sia il prodotto scalare che il vettoriale.
* Verifica ortogonalità se il prodotto scalare è zero.

```python
import numpy as np
```

```python
v1 = np.random.uniform(0, 11, 3)
v2 = np.random.uniform(0, 11, 3)
```

```python
prodott_scal = np.dot(v1, v2)
prodott_scal
```

```python
prod_vett = np.cross(v1, v2)
prod_vett
```

```python
if prodott_scal == 0:
    print("I vettori sono ortogonali")
else:
    print("I vettori non sono ortogonali")
```
