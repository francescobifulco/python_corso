## Esercizio 10
* Imposta un seed casuale (es. `np.random.seed(42)`).
* Genera un array 4×5 di numeri casuali tra -1 e 1.
* Calcola i valori assoluti e trova la posizione del massimo valore assoluto.

```python
import numpy as np
```

```python
np.random.seed(50)
```

```python
v1 = np.random.uniform(-1, 1, (4,5))
v1
```

```python
valore_asso = np.abs(v1)
valore_asso
```

```python
max_assoluto = np.max(valore_asso)
max_assoluto
```

```python
posizione = np.unravel_index(np.argmax(valore_asso), valore_asso.shape)
posizione
```
