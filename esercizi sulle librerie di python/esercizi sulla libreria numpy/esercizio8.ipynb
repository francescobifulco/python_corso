## Esercizio 8 
* Crea un array di float casuali (10 elementi).
* Converti l’array in interi e confronta le differenze di valore.
* Calcola l’errore medio assoluto tra i due array.

```python
import numpy as np
```

```python
v1 = np.random.rand(10)
v1
```

```python
v1_int = v1.astype('i4')
v1_int
```

```python
differenza = np.subtract(v1, v1_int)
differenza
```

```python
errore_medio_assoluto = np.mean(np.abs(differenza))
errore_medio_assoluto
```
