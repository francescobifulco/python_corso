## Esericizio 4
Genera 30 numeri equispaziati tra 0 e 10.
Rimodella l’array in una matrice 5×6.
Calcola la media e la deviazione standard per ogni colonna.

```python
import numpy as np
```

```python
v1 = np.linspace(0, 11, 30)
v1
```

```python
v1 = v1.reshape((5,6))
v1
```

```python
media = np.mean(v1, axis=0)
media
```

```python
deviazione_standard = np.std(v1, axis=0)
deviazione_standard
```
