## Esercizio 17
* Genera un array 100×3 di valori casuali (simulando 100 osservazioni di 3 variabili).
* Normalizza ciascuna colonna tra 0 e 1.
* Calcola la media e la deviazione standard delle colonne normalizzate.

```python
import numpy as np
```

```python
v1 = np.random.rand(100, 3)
v1
```

```python
min_vals = v1.min(axis=0)
max_vals = v1.max(axis=0)
```

```python
v_norm = (v1 - min_vals) / (max_vals - min_vals)
v_norm
```

```python
mean_vals = v_norm.mean(axis=0)
std_vals  = v_norm.std(axis=0)
mean_vals, std_vals
```
