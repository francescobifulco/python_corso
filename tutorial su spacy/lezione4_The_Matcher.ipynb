Il Matcher è uno strumento efficiente per trovare pattern basati su attributi linguistici (POS, lemma, forma del testo) piuttosto che solo su espressioni regolari.

```python
from spacy.matcher import Matcher
import spacy
```

```python
nlp = spacy.load("it_core_news_sm")
```

```python
testo = "L'Etna, il vulcano attivo più alto d'Europa, si trova in Sicilia."
```

```python
doc = nlp(testo)
```

```python
# Inizializza il Matcher con il Vocab condiviso
matcher = Matcher(nlp.vocab)
```

```python
# Definizione di un Pattern (Adjective + Noun)
# Trova la sequenza di un NOUN seguito da un ADJ
pattern_noun_adj = [{"POS": "NOUN"}, {"POS": "ADJ"}]
```

```python
# Aggiunta del pattern al Matcher con un ID
matcher.add("AGGETTIVO_NOME", [pattern_noun_adj])
```

```python
# Applicazione del Matcher al Doc
matches = matcher(doc)
```

```python
print("\n--- Matcher (Regole Linguistiche) ---")
for match_id, start, end in matches:
    # Ottenere la stringa del match ID
    string_id = nlp.vocab.strings[match_id]
    # Ottenere lo Span corrispondente ai token trovati
    span = doc[start:end]
    print(f"ID Match: {string_id} | Span Trovato: {span.text}")

# Esempio trovato: 'attivo più alto' (se ADJ+ADJ+NOUN), 'vulcano attivo'
```

```python
from spacy.matcher import PhraseMatcher
```

```python
# Inizializza il PhraseMatcher (più veloce del Matcher per frasi esatte)
phrase_matcher = PhraseMatcher(nlp.vocab)
```

```python
# Frase da cercare
frase_cercata = "vulcano attivo più alto d'Europa"
```

```python
# Creazione di un Doc per la frase (necessario per PhraseMatcher)
frase_doc = nlp(frase_cercata)
```

```python
# Aggiunta del pattern (è l'oggetto Doc della frase)
phrase_matcher.add("VULCANO_IMPORTANTE", [frase_doc])
```

```python
# Applicazione al Doc originale
phrase_matches = phrase_matcher(doc)
```

```python
print("\n--- PhraseMatcher (Frasi Esatte) ---")
for match_id, start, end in phrase_matches:
    span = doc[start:end]
    print(f"Frase Trovata: {span.text}")
```
