```python
import spacy
```

```python
nlp = spacy.load("it_core_news_sm")
```

```python
testo = "L'Etna, il vulcano attivo più alto d'Europa, si trova in Sicilia."
```

```python
doc = nlp(testo)
```

doc → è l’oggetto Doc da cui il testo proviene

start → indice del primo token incluso nello span

end → indice del token successivo all’ultimo (come negli slicing Python)

label (opzionale) → etichetta che puoi associare allo span (ad esempio una categoria semantica o NER)

In un Doc o in una frase (Span che rappresenta una frase), il .root è il verbo principale o il nucleo sintattico.

In uno span arbitrario, .root è il token grammaticalmente dominante all’interno di quello span.

Ogni token in spaCy ha un attributo .head, che è un altro Token da cui dipende grammaticalmente.

| Concetto        | Significato                        | Analogia             |
| --------------- | ---------------------------------- | -------------------- |
| **`.root`**     | Radice della frase o dello span    | Il **capo famiglia** |
| **`.head`**     | Il padre (governatore) di un token | “da chi dipende”     |
| **`.children`** | Tutti i figli di un token          | “chi dipende da lui” |


```python
# Creazione manuale di un oggetto Span
# Si usa la notazione di slicing Python [start_index : end_index]
# Lo Span da ',il vulcano attivo...'
span_vulcano = doc[2:6]

print("--- L'Oggetto Span ---")
print(f"Testo dello Span: {span_vulcano.text}")
print(f"Il Root dello Span (testa grammaticale): {span_vulcano.root.text}")
print(f"Genitori dell'intero Span (solo se Span è una singola entità o radice): {span_vulcano.root.head.text}")
```

In spaCy, l’oggetto Vocab è una delle componenti centrali dell’intero sistema linguistico.
Rappresenta il vocabolario condiviso — cioè il magazzino dove spaCy tiene tutte le informazioni linguistiche su parole, lemmi, vettori, morfologia e simboli.

in spaCy, gli hash id (o più precisamente string hash IDs) sono numeri interi univoci che rappresentano stringhe di testo in modo efficiente.

Sono una parte fondamentale del sistema interno di spaCy, usata per rendere il lavoro con grandi quantità di testo molto più veloce e leggero in memoria.

```python
# Accedere all'oggetto Vocab della pipeline
vocab = nlp.vocab

# La stringa di testo "Etna"
etna_string = "Etna"

# 1. Ottenere l'hash ID (l'identificatore numerico) della stringa
etna_hash = vocab.strings[etna_string]
print(f"\n--- Vocab e StringStore ---")
print(f"Stringa: {etna_string} | Hash ID: {etna_hash}")

# 2. Ricostruire la stringa partendo dall'Hash ID
etna_string_from_hash = vocab.strings[etna_hash]
print(f"Stringa recuperata dall'Hash: {etna_string_from_hash}")

# Nota: i Token, gli Span e le Entità memorizzano solo l'ID hash 
# per risparmiare memoria.
```
