```python
import spacy
from spacy.language import Language
```

```python
nlp = spacy.load("it_core_news_sm")
```

```python
testo = "L'Etna, il vulcano attivo più alto d'Europa, si trova in Sicilia."
```

```python
doc = nlp(testo)
```

```python
# Decoratore per registrare la funzione come componente della pipeline
@Language.component("conteggio_vulcani")
def conteggio_vulcani_component(doc):
    # Logica del componente: conteggia quante volte
    # appare la parola 'vulcano' (o lemma)
    conteggio = 0
    for token in doc:
        if token.lemma_ == "vulcano":
            conteggio += 1

    # Utilizzo degli Attribute Extensions (vedi 5.2) 
    # per salvare il risultato nel Doc
    doc._.conteggio_vulcani = conteggio
    return doc
```

```python
# Registrazione dell'estensione (vedi 5.2)
# Se non fosse già stata registrata, darebbe errore!
# Registrazione dell'estensione per l'oggetto Doc
if not spacy.tokens.Doc.has_extension("conteggio_vulcani"):
    spacy.tokens.Doc.set_extension("conteggio_vulcani", default=0)
```

```python
# Aggiunta del componente alla pipeline (dopo il NER)
# 'last=True' lo aggiunge come ultimo componente
nlp.add_pipe("conteggio_vulcani", last=True)
```

```python
# Test con il Doc originale
doc_nuovo = nlp(doc.text)
```

```python
print("\n--- Custom Pipeline Component ---")
# Accesso all'attributo personalizzato
print(f"Numero di volte che appare 'vulcano' nel Doc: {doc_nuovo._.conteggio_vulcani}")
```

```python
# Rimozione del componente (per non interferire con il training)
nlp.remove_pipe("conteggio_vulcani")
```

```python
# Registrazione dell'estensione per l'oggetto Token
if not spacy.tokens.Token.has_extension("is_maiuscola"):
    # Setter: funzione per calcolare il valore 
    # (e.g., True se inizia con maiuscola)
    def is_maiuscola_getter(token):
        return token.text[0].isupper()

    # Registra l'estensione 'is_maiuscola' sul Token
    spacy.tokens.Token.set_extension("is_maiuscola", getter=is_maiuscola_getter)

print("\n--- Attribute Extensions (Token) ---")
for token in doc:
    # Accesso all'attributo personalizzato
    print(f"Token: {token.text:<10} | Inizia con maiuscola?: {token._.is_maiuscola}")
```
