```python
# Il Dependency Parsing mostra le relazioni 
# grammaticali tra i token, cruciali per 
# comprendere la struttura della frase.
import spacy
from spacy import displacy
```

```python
nlp = spacy.load("it_core_news_sm")
```

```python
testo = "L'Etna, il vulcano attivo più alto d'Europa, si trova in Sicilia."
```

```python
doc = nlp(testo)
```

```python
print("--- Analisi delle Dipendenze ---")
for token in doc:
    # 'token.dep_' è il tipo di relazione di dipendenza
    # 'token.head.text' è il token a cui è collegato (il suo genitore)
    print(f"Token: {token.text:<10} | "
          f"Dipendenza: {token.dep_:<15} | "
          f"Genitore (Head): {token.head.text}")
```

# Etichetta (Label),Descrizione
1. LOC,"Luoghi (città, regioni, catene montuose, ecc.)"
2. PER,Persone nominate.
3. ORG,"Organizzazioni, aziende, istituzioni."

```python
print("\n--- Riconoscimento di Entità (NER) ---")

# L'oggetto Doc ha una proprietà 'ents' che contiene 
# tutte le entità riconosciute
for entita in doc.ents:
    # 'entita.text' è il testo dell'entità
    # 'entita.label_' è l'etichetta (tipo) dell'entità
    print(f"Entità: {entita.text:<15} | Tipo: {entita.label_:<10}")
```
