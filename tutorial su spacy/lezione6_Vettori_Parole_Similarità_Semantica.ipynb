```python
# Installazione del modello medio (md) che include i vettori
# Questo modello è più grande e lento, ma ha i vettori di 
# parole necessari per la similarità.
```

```python
import spacy
```

```python
# Caricamento del modello medio
nlp_md = spacy.load("it_core_news_md")
```

```python
# Testo di esempio
testo_sim = "Il cane beve l'acqua. La Sicilia è un'isola."
doc_sim = nlp_md(testo_sim)
```

ogni token, frase o documento può avere un vettore numerico (embedding), che rappresenta il suo significato in uno spazio continuo di dimensione fissa (es. 300 dimensioni).
Questi vettori servono per confrontare parole simili o misurare similarità semantiche.

```python
# I token ora hanno un vettore numerico associato 
# (di 300 dimensioni nel modello md)
token_cane = doc_sim[1] # 'cane'
print("--- 6.1. Vettori di Parole ---")
print(f"Token: {token_cane.text}")
print(f"Vettore (prime 5 dimensioni): {token_cane.vector[:5]}")
print(f"Lunghezza del vettore: {len(token_cane.vector)}")
```

```python
# Controllo se il token ha un vettore
print(f"Il token 'cane' ha un vettore?: {token_cane.has_vector}")
```

```python
# Calcola la similarità tra due token (valore tra 0 e 1, 1 = identici)
token_acqua = doc_sim[3] # 'acqua'
token_isola = doc_sim[8] # 'isola'
```

```python
# Similitudine tra 'cane' e 'acqua'
sim_cane_acqua = token_cane.similarity(token_acqua)
print(f"\nSimilarità tra 'cane' e 'acqua': {sim_cane_acqua:.4f}")
```

```python
sim_acqua_isola = token_acqua.similarity(token_isola)
print(f"\nSimilarità tra 'acqua' e 'isola': {sim_acqua_isola:.4f}")
```

```python
# Similitudine tra due Span (e.g., Doc interi o frasi)
# Similarità tra la prima frase ("Il cane beve l'acqua.") 
# e la seconda ("La Sicilia è un'isola.")
frase1 = doc_sim[:4]
frase2 = doc_sim[5:]
sim_frasi = frase1.similarity(frase2)
print(f"Similarità tra le frasi: {sim_frasi:.4f}")
```

```python
sim_acqua_acqua = token_acqua.similarity(token_acqua)
print(f"\nSimilarità tra 'acqua' e 'acqua': {sim_acqua_acqua:.4f}")
```
