```python
import spacy 
```

```python
# Caricamento della pipeline NLP (Natural Language Processing)
# 'spacy.load()' carica il modello italiano scaricato
nlp = spacy.load("it_core_news_sm")
```

```python
# Testo di esempio
testo = "L'Etna, il vulcano attivo più alto d'Europa, si trova in Sicilia."
```

```python
# Creazione dell'oggetto Doc
# L'oggetto 'doc' viene creato passando il testo alla pipeline 'nlp'.
# Questo esegue tutte le operazioni di elaborazione 
# definite nel modello (tokenizzazione, POS, NER, ecc.).
doc = nlp(testo)
```

```python
# Stampa dell'oggetto Doc
doc
```

```python
# Iterazione su ogni token (unità) nell'oggetto Doc
print("--- Tokenizzazione ---")
for token in doc:
    # 'token.text' è il testo effettivo del token
    # 'token.i' è l'indice del token all'interno del Doc
    print(f"Token: {token.text}\t Indice: {token.i}")
```

```python
print("\n--- Annotazioni dei Token ---")
for token in doc:
    print(f"Testo: {token.text:<10} | " # Il testo esatto del token.
          f"POS (Categoria Grammaticale): {token.pos_:<5} | " # La parte del discorso (Part-of-Speech), e.g., NOUN, VERB, ADJ.
          f"Lemma (Forma base): {token.lemma_:<10} | " # La forma base o dizionario della parola
          f"Stopword?: {token.is_stop}") # True se il token è una stopword (parola comune ignorabile).
```
