```python
import spacy
```

```python
nlp = spacy.load("it_core_news_sm")
```

```python
frase = 'Il gatto nero dorme sul divano vicino alla finestra.'
```

```python
doc = nlp(frase)
doc
```

```python
print("--- Tokenizzazione ---")
for token in doc:
    print(f"Token: {token.text}\t Indice: {token.i}")
```

```python
print("--- lemmatizzazione ---")
for token in doc:
    print(f"Testo: {token.text}\tLemma: {token.lemma_}\tIndice: {token.i}")
```

```python
print("--- Part-of-Speech Tagging ---")
for token in doc:
    print(f"Testo: {token.text:9}\tPOS {token.pos_:9}\tIndice: {token.i}")
```

```python
print("\n--- Annotazioni dei Token ---")
for token in doc:
    print(f"Testo: {token.text:<10} | "
          f"POS (Categoria Grammaticale): {token.pos_:<5} | " 
          f"Lemma (Forma base): {token.lemma_:<10} | ")
```

```python
articolo = """L’importanza del riciclo nella vita quotidiana
Negli ultimi anni, la consapevolezza ambientale è aumentata in modo significativo.
Sempre più persone cercano di ridurre i rifiuti e di riutilizzare i materiali, 
contribuendo così alla protezione del pianeta.
Le scuole e le aziende promuovono campagne di sensibilizzazione per incoraggiare 
comportamenti sostenibili.
Tuttavia, non tutti conoscono le corrette pratiche di raccolta differenziata, 
e spesso gli errori di smaltimento riducono l’efficacia del riciclo.
Educare i cittadini e migliorare i sistemi di gestione dei rifiuti 
resta una priorità per costruire un futuro più pulito e responsabile."""
```

```python
doc_articolo = nlp(articolo)
doc_articolo
```

```python
print("\n--- Riconoscimento di Entità (NER) ---")

for entita in doc_articolo.ents:
    print(f"Entità: {entita.text:<15} | Tipo: {entita.label_:<10}")
```

```python
from collections import Counter

entita_labels = [ent.label_ for ent in doc_articolo.ents]
conteggio = Counter(entita_labels)

print("\n--- Frequenza delle Entità ---")
for tipo, count in conteggio.items():
    print(f"{tipo}: {count}")
```
