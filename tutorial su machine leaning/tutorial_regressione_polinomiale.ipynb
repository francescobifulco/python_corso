```python
import numpy
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
x = [1,2,3,5,6,7,8,9,10,12,13,14,15,16,18,19,21,22]
y = [100,90,80,60,60,55,60,65,70,70,75,76,78,79,90,99,99,100]
```

```python
# Creare un modello polinomiale di grado 3
# numpy.polyfit(x, y, degree) trova i coefficienti del polinomio
# numpy.poly1d(coefficients) crea una funzione polinomiale
mymodel = numpy.poly1d(numpy.polyfit(x, y, 3))
```

```python
# Creare una sequenza di punti x per disegnare la linea del modello
myline = numpy.linspace(1, 22, 100) # 100 punti tra 1 e 22
```

```python
# Disegnare i dati originali e la linea del modello
plt.scatter(x, y)
plt.plot(myline, mymodel(myline), color='red')
plt.show()
```

```python
# Calcolare l'R-quadro
print(r2_score(y, mymodel(x)))
```

```python
# Prevedere il valore per un nuovo punto X (es. x=17)
print(mymodel(17))
```
