### Creare una classe padre
Qualsiasi classe può essere una classe padre, quindi la sintassi è la stessa della creazione di qualsiasi altra classe:

```python
class Persona:
    def __init__(self, nome, cognome):
        self.nome = nome
        self.cognome = cognome
    
    def stampa(self):
        print(self.nome, self.cognome)

x = Persona('franci', 'bifu')
x.stampa()
```

### Creare una classe figlia
Per creare una classe che eredita la funzionalità da un'altra classe, inviare la classe padre come parametro durante la creazione della classe figlia:

```python
class Studente(Persona):
  pass
```

```python
x = Studente('dario', 'esposito')
x.stampa()
```

### Aggiungere la funzione __init__()
Finora abbiamo creato una classe figlia che eredita le proprietà e i metodi dalla classe madre.

Vogliamo aggiungere la __init__()funzione alla classe figlia (invece della passparola chiave).

```python
class Studente1(Persona):
    def __init__(self, nome, cognome):
        super().__init__(nome, cognome)

x = Studente1('ciro', 'rossi')
x.stampa()
```

### Utilizzare la funzione super()
Python ha anche una super()funzione che fa sì che la classe figlia erediti tutti i metodi e le proprietà dalla classe madre:

```python
class Persona:
    def __init__(self, nome, cognome):
        self.nome = nome
        self.cognome = cognome
    
    def stampa(self):
        print(self.nome, self.cognome)

class Studente(Persona):
    def __init__(self, nome, cognome):
        super().__init__(nome, cognome)

x = Studente('luigi', 'scogliamiglio')
x.stampa()
```

#### Aggiungi proprietà


```python
class Studente(Persona):
    def __init__(self, nome, cognome, anno_nascita):
        super().__init__(nome, cognome)
        self.anno_nascita = anno_nascita
    
    def stampa(self):
        print(self.nome, self.cognome, self.anno_nascita)

x = Studente('luca', 'casol', 2000)
x.stampa()
```

### Aggiungi metodi

```python
class Studente(Persona):
    def __init__(self, nome, cognome, anno_nascita):
        super().__init__(nome, cognome)
        self.anno_nascita = anno_nascita
    
    def benvenuto(self):
        print(f'Benvenuto a {self.nome} {self.cognome} e nato il  {self.anno_nascita}')
    
x = Studente('lucia', 'casol', 2002)
x.benvenuto()
```
