## Polimorfismo di funzione
Un esempio di funzione Python che può essere utilizzata su oggetti diversi è la len()funzione.

## Corda
Per le stringhe len()restituisce il numero di caratteri:

```python
x = 'Ciao mondo!'

len(x)
```

## Tupla
Per le tuple len()restituisce il numero di elementi nella tupla:

```python
tupla = ('mela', 'banana', 'pera', 'ananas')

len(tupla)
```

## Dizionario
Per i dizionari len()restituisce il numero di coppie chiave/valore nel dizionario:

```python
Persona = {
    'nome': 'ciro',
    'cognome': 'esposito',
    'eta': '34'
}

len(Persona)
```

## Polimorfismo di classe
Il polimorfismo viene spesso utilizzato nei metodi di classe, dove possiamo avere più classi con lo stesso nome di metodo.

Ad esempio, supponiamo di avere tre classi: Car, Boat, e Plane, e che tutte abbiano un metodo chiamato move():

```python
class Car:
  def __init__(self, marchio, modello):
    self.marchio = marchio
    self.modello = modello

  def move(self):
    print("Drive!")

class Boat:
  def __init__(self, marchio, modello):
    self.marchio = marchio
    self.modello = modello

  def move(self):
    print("Sail!")

class Plane:
  def __init__(self, marchio, modello):
    self.marchio = marchio
    self.modello = modello

  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang")       
boat1 = Boat("Ibiza", "Touring 20") 
plane1 = Plane("Boeing", "747")     

for x in (car1, boat1, plane1):
  x.move()
```

## Polimorfismo delle classi di ereditarietà
E le classi con classi figlie con lo stesso nome? Possiamo usare il polimorfismo in questo caso?

Sì. Se utilizziamo l'esempio precedente e creiamo una classe padre chiamata Vehicle, e creiamo Car, Boat, Plane classi figlio di Vehicle, la classe figlio eredita i Vehiclemetodi, ma può sovrascriverli:

```python
class Veicolo:
  def __init__(self, marchio, modello):
    self.marchio = marchio
    self.modello = modello

  def move(self):
    print("Move!")

class Car(Veicolo):
  pass

class Boat(Veicolo):
  def move(self):
    print("Sail!")

class Plane(Veicolo):
  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang") 
boat1 = Boat("Ibiza", "Touring 20") 
plane1 = Plane("Boeing", "747") 

for x in (car1, boat1, plane1):
  print(x.marchio)
  print(x.modello)
  x.move()
```
