### Iteratori Python
Un iteratore è un oggetto che contiene un numero numerabile di valori.

Un iteratore è un oggetto su cui è possibile eseguire un'iterazione, ovvero è possibile attraversare tutti i valori.

Tecnicamente, in Python, un iteratore è un oggetto che implementa il protocollo iteratore, che consiste nei metodi __iter__() e __next__().

### Iteratore vs Iterabile
Liste, tuple, dizionari e insiemi sono tutti oggetti iterabili. Sono contenitori iterabili da cui è possibile ottenere un iteratore.

Tutti questi oggetti hanno un iter() un metodo che viene utilizzato per ottenere un iteratore:

```python
tupla = ("mele", "banana", "pera")
myit = iter(tupla)

print(next(myit))
print(next(myit))
print(next(myit))
```

```python
stringa = "banana"
myit = iter(stringa)

print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

## Esecuzione di un ciclo attraverso un iteratore

```python
for x in tupla:
    print(x)
```

```python
for x in stringa:
    print(x)
```

## Creare un iteratore
Per creare un oggetto/classe come iteratore è necessario implementare i metodi __iter__()e __next__()per l'oggetto.

Come hai imparato nel capitolo Classi/Oggetti Python , tutte le classi hanno una funzione chiamata __init__(), che consente di eseguire alcune inizializzazioni quando l'oggetto viene creato.

Il __iter__()metodo funziona in modo simile: è possibile eseguire operazioni (inizializzazione, ecc.), ma è sempre necessario restituire l'oggetto iteratore stesso.

Il __next__()metodo consente anche di eseguire operazioni e deve restituire l'elemento successivo nella sequenza.

```python
class Numeri:
  def __iter__(self):
    self.a = 1
    return self

  def __next__(self):
    x = self.a
    self.a += 1
    return x

classNumeri = Numeri()
myiter = iter(classNumeri)

print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
```

## Interrompi iterazione
L'esempio sopra continuerebbe all'infinito se ci fossero sufficienti istruzioni next() o se venisse utilizzato in un ciclo.

Per evitare che l'iterazione continui all'infinito, possiamo usare l' StopIterationistruzione.

Nel __next__()metodo possiamo aggiungere una condizione di terminazione per generare un errore se l'iterazione viene eseguita un numero specificato di volte:

```python
class Numeri:
  def __iter__(self):
    self.a = 1
    return self

  def __next__(self):
      if self.a <= 10:
          x = self.a
          self.a += 1
          return x
      else:
          raise StopIteration

classNumeri = Numeri()
myiter = iter(classNumeri)

for x in myiter:
  print(x)
```
