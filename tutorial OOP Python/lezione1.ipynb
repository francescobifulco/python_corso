### Crea una classe
Per creare una classe, utilizzare la parola chiave class:

```python
class MiaClasse:
  x = 5

print(MiaClasse)
```

### Crea oggetto
Ora possiamo usare la classe denominata MyClass per creare oggetti:

```python
p1 = MiaClasse()
print(p1.x)
```

### Il metodo __init__()
Gli esempi sopra riportati sono classi e oggetti nella loro forma più semplice e non sono realmente utili nelle applicazioni della vita reale.

Per comprendere il significato delle classi dobbiamo comprendere il metodo incorporato __init__() .

Tutte le classi hanno un metodo chiamato __init__(), che viene sempre eseguito quando la classe viene avviata.

Utilizzare il __init__()metodo per assegnare valori alle proprietà dell'oggetto o altre operazioni necessarie da eseguire durante la creazione dell'oggetto:

```python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

p1 = Persona("John", 36)

print(p1.nome)
print(p1.eta)
```

### Il metodo __str__()
Il __str__()metodo controlla cosa deve essere restituito quando l'oggetto classe è rappresentato come stringa.

Se il __str__()metodo non è impostato, viene restituita la rappresentazione stringa dell'oggetto:

```python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta
    
    def __str__(self):
        return f'{self.nome} {self.eta}'

p1 = Persona("John", 36)
p1
```

### Creare metodi
È possibile creare metodi personalizzati all'interno degli oggetti. I metodi negli oggetti sono funzioni che appartengono all'oggetto.

Creiamo un metodo nella classe Person:



```python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta
    
    def mioMetodo(self):
        print(f'Ciao il mio nome e {self.nome}')

p1 = Persona("John", 36)
p1.mioMetodo()
```

### Il parametro self
Il self parametro è un riferimento all'istanza corrente della classe e viene utilizzato per accedere alle variabili che appartengono alla classe.

Non è necessario che abbia un nome self, puoi chiamarlo come preferisci, ma deve essere il primo parametro di qualsiasi funzione nella classe:

```python
class Person:
  def __init__(mysillyobject, name, age):
    mysillyobject.name = name
    mysillyobject.age = age

  def myfunc(abc):
    print("Hello my name is " + abc.name)

p1 = Person("John", 36)
p1.myfunc()
```

### Modifica le proprietà dell'oggetto
È possibile modificare le proprietà degli oggetti in questo modo:

```python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta
    
    def mioMetodo(self):
        print(f'Ciao il mio nome e {self.nome}')

p1 = Persona("John", 36)
p1.eta = 49

p1.eta
```

### Elimina le proprietà dell'oggetto
È possibile eliminare le proprietà degli oggetti utilizzando la delparola chiave:

```python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta
    
    def mioMetodo(self):
        print(f'Ciao il mio nome e {self.nome}')

p1 = Persona("John", 36)
del p1.eta

p1.eta
```

### Elimina oggetti
È possibile eliminare gli oggetti utilizzando la delparola chiave:

```python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta
    
    def mioMetodo(self):
        print(f'Ciao il mio nome e {self.nome}')

p1 = Persona("John", 36)
del p1

print(p1)
```
