# Esercizio 5.1: Confronta prestazioni tra Decision Tree e Random Forest
# usando il dataset Iris con cross-validation

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingRegressor
import time
from sklearn.ensemble import AdaBoostClassifier
```

```python

iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

clf = DecisionTreeClassifier(random_state=42)
rf = RandomForestClassifier(n_estimators=100, random_state=42)

scoresRf = cross_val_score(rf, X, y, cv=5)
print("Accuracy media RF:", scoresRf.mean())

scores = cross_val_score(clf, X, y, cv=5)
print("Accuracy media:", scores.mean())
```

# Esercizio 5.2: Confronta tre metodi ensemble sul dataset Breast Cancer
# Analizza accuracy e tempi di training
Confronta Random Forest, Gradient Boosting e AdaBoost.

```python
coseBrutte = load_breast_cancer()
X, y = coseBrutte.data, coseBrutte.target

```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42)
start = time.time()
rf.fit(X_train, y_train)
end = time.time()
print("tempo di esecuzione", end-start)

# Valuta il modello
accuracy = rf.score(X_test, y_test)
print("Accuracy:", accuracy)
```

```python
gb = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42)

start = time.time()
gb.fit(X_train, y_train)
end = time.time()
print("tempo di esecuzione", end-start)

# Valuta il modello
accuracy = gb.score(X_test, y_test)
print("Accuracy:", accuracy)
```

```python
adaboost = AdaBoostClassifier(n_estimators=50, random_state=42)
start = time.time()
adaboost.fit(X_train, y_train)

end = time.time()
print("tempo di esecuzione", end-start)

# Valuta il modello
accuracy = adaboost.score(X_test, y_test)
print("Accuracy:", accuracy)
```

# Esercizio 5.3: Ottimizza Gradient Boosting con Grid Search
# Focus su learning_rate, n_estimators, max_depth
Ottimizza un Gradient Boosting con Grid Search per massimizzare le prestazioni.

```python
gb = GradientBoostingRegressor(random_state=42)

param_grid = {'learning_rate' : [1, 0.5, 0.25, 0.1, 0.05, 0.01], 'n_estimators': [50, 100, 150], 'max_depth': [2,3,4]}

# learning_rates = [1, 0.5, 0.25, 0.1, 0.05, 0.01]
# n_estimators = [1, 2, 4, 8, 16, 32, 64, 100, 200]
# np.linspace(1, 32, 32, endpoint=True)

grid_search = GridSearchCV(estimator=gb, param_grid=param_grid, cv=5)
grid_search.fit(X_train, y_train)

print("Migliori parametri:", grid_search.best_params_)

# Valuta il modello
accuracy = grid_search.score(X_test, y_test)
print("Accuracy:", accuracy)
```

```python
gb = GradientBoostingRegressor(random_state=42)

param_grid = {'learning_rate' : [1, 0.5, 0.25, 0.1, 0.05, 0.01], 'n_estimators': [50, 100, 150], 'max_depth': [2,3,4]}

# learning_rates = [1, 0.5, 0.25, 0.1, 0.05, 0.01]
# n_estimators = [1, 2, 4, 8, 16, 32, 64, 100, 200]
# np.linspace(1, 32, 32, endpoint=True)

grid_search = GridSearchCV(estimator=gb, param_grid=param_grid, cv=5)
scores = cross_val_score(grid_search, X, y, cv=5)
print("Accuracy media grid search:", scores.mean())

print("Migliori parametri:", grid_search.best_params_)
```
