```python
#len(): Data la stringa "Programmazione Python",
#calcola e stampa la sua lunghezza.
testo = "Programmazione Python"
len(testo)
```

```python
#upper() e lower(): Data la stringa "Ciao Mondo", 
#convertila prima tutta in maiuscolo e poi tutta in minuscolo. 
# Stampa entrambi i risultati.
testo = "Ciao Mondo"
testo.upper(), testo.lower()
```

```python
#capitalize(): Data la stringa "benvenuti in python", 
#capitalizza solo la prima lettera della frase.
testo = "benvenuti in python"
testo.capitalize()
```

```python
#title(): Data la stringa "impara python con esercizi", 
#convertila in modo che ogni parola inizi con una lettera 
#maiuscola.
testo =  "impara python con esercizi"
testo.title()
```

```python
#count(): Data la stringa "banana", conta quante volte appare la lettera 'a'.
parola = "banana"
parola.count('a')
```

```python
#find(): Data la stringa "Python è un linguaggio potente.", 
#trova l'indice della prima occorrenza della parola 
# "linguaggio". Cosa succede se cerchi una parola che non c'è?
testo = "Python è un linguaggio potente."
testo.find('linguaggio'), testo.find('ciao')
```

```python
#rfind(): Data la stringa "Mississippi", trova l'indice 
#dell'ultima occorrenza della lettera 's'.
parola = "Mississippi"
parola.rfind('s')
```

```python
#index(): Data la stringa "programmazione", trova l'indice 
#della prima occorrenza della lettera 'o'. Cosa succede se 
#cerchi una lettera che non c'è? (Nota la differenza con 
#find()).
parola = "programmazione"
parola.index('o') #parola.index('q')
```

```python
#rindex(): Data la stringa "abracadabra", trova l'indice 
#dell'ultima occorrenza della lettera 'a'.
parola = "abracadabra"
parola.rindex('a')
```

```python
#replace(): Data la stringa "Mi piace il caffè", 
#sostituisci "caffè" con "tè".
testo = "Mi piace il caffè"
testo.replace('caffè','tè')
```

```python
#strip(): Data la stringa "   Spazi intorno   ", rimuovi 
#gli spazi bianchi all'inizio e alla fine.
testo = "   Spazi intorno   "
testo.strip()
```

```python
#lstrip() e rstrip(): Data la stringa 
# "###Testo con cancelletti###", rimuovi solo i cancelletti 
# iniziali e poi solo quelli finali.
testo = "###Testo con cancelletti###"
print(testo.lstrip('#'))  # Rimuove i cancelletti iniziali
print(testo.rstrip('#'))  # Rimuove i cancelletti finali
```
