### Creazione di una tabella

Per creare una tabella in MySQL, utilizzare la dichiarazione "CREATE TABLE".
Assicurarsi di definire il nome del database quando si crea la connessione

```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="databese_python"
)

mycursor = mydb.cursor()

mycursor.execute("CREATE TABLE Prodotti (Nome_prodotto VARCHAR(255))")
```

### Controlla se la tabella esiste

Puoi verificare se una tabella esiste elencando tutte le tabelle nel tuo database con l'istruzione "TABLESSHOW"

```python
mycursor = mydb.cursor()

mycursor.execute("SHOW TABLES")

for x in mycursor:
  print(f'tutte le mie tabelle nel DataBase: {x}') 
```

### La chiave principale

Quando si crea una tabella, si dovrebbe anche creare una colonna con una chiave unica per ogni - Registrazione.

Questo può essere fatto definendo una chiave PRIMARY.

Usiamo la dichiarazione "INT AUTO-INCREMENT PRIMARY KEY" che inserirà un Numero unico per ogni record. A partire da 1, e aumentato di uno per ciascuno - Registrazione.

```python
mycursor.execute("CREATE TABLE Categorie (id_Categorie INT AUTO_INCREMENT PRIMARY KEY, nome_categorie VARCHAR(255))") 
```

Se la tabella esiste già, utilizzare la parola chiave ALTER TABLE

```python
mycursor.execute("ALTER TABLE Utente ADD COLUMN id_Utenti INT AUTO_INCREMENT PRIMARY KEY")
```
