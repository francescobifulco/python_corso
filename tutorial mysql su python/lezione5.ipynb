### Seleziona da un tabella

Per selezionare da una tabella in MySQL, utilizzare l'istruzione "SELECT":

```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="databese_python"
)

mycursor = mydb.cursor()
```

```python
mycursor.execute("SELECT * FROM utente")

myresult = mycursor.fetchall()

for x in myresult:
  print(f'tutte le righe nella tabella utente: {x}')
```

### La scelta delle colonne

Per selezionare solo alcune delle colonne in una tabella, utilizzare l'istruzione "SELECT" Seguono i nomi della colonna

```python
mycursor.execute("SELECT nome FROM utente")

myresult = mycursor.fetchall()

for x in myresult:
  print(f'tutte le righe della colonna nome della tebella utente: {x}') 
```

### Usando il metodo di fetchone()

Se sei interessato solo a una riga, puoi usare il fetchone()
Il più grande fetchone() Il metodo restituirà la prima riga di Il risultato

```python
mycursor.execute("SELECT * FROM utente")

myresult = mycursor.fetchone()

print(myresult) 
```
