### Creazione di un database

Per creare un database in MySQL, utilizzare l'istruzione "CREATE DATABASE"

```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root"
)

mycursor = mydb.cursor()

mycursor.execute("CREATE DATABASE Databese_Python")
```

### Controllare se il database esiste

È possibile verificare se un database esiste elencando tutti i database nel sistema per Utilizzando la dichiarazione "SHOW DATABASES"

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root"
)

mycursor = mydb.cursor()

mycursor.execute("SHOW DATABASES")

for x in mycursor:
  print(f'tutti i miei DataBase: {x}') 
```

Oppure puoi provare ad accedere al database quando effettui la connessione

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="databese_python"
)
```
