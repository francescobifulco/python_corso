### Seleziona con un filtro

Quando si selezionano i record da una tabella, è possibile filtrare la selezione utilizzando La dichiarazione "Where"

```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="databese_python"
)

mycursor = mydb.cursor()
```

```python
sql = "SELECT * FROM utente WHERE cognome =''"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
  print(x)
```
