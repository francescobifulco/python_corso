### Inserire nella tabella

Per riempire una tabella in MySQL, utilizzare la dichiarazione "INSERT INTO".

```python
import mysql.connector
```

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="databese_python"
)

mycursor = mydb.cursor()
```

```python
sql = "INSERT INTO Utente (nome, cognome) VALUES (%s, %s)"
val = ("federico", "rossi")
mycursor.execute(sql, val)

mydb.commit()

print(f"{mycursor.rowcount} record inserted.")
```

## Inserire più righe

Per inserire più righe in una tabella, utilizzare il executemany() Un metodo.

Il secondo parametro del executemany() Il metodo di Un elenco di tuples, contenente i dati che si desidera inserire

```python
sql = "INSERT INTO Utente (nome, cognome) VALUES (%s, %s)"
val = [
  ('Peter', 'Lowstreet'),
  ('Amy', 'Apple st'),
  ('Hannah', 'Mountain'),
  ('Michael', 'Valley'),
  ('Sandy', 'Ocean blvd'),
  ('Betty', 'Green Grass'),
  ('Richard', 'Sky st'),
  ('Susan', 'One way'),
  ('Vicky', 'Yellow Garden'),
  ('Ben', 'Park Lane'),
  ('William', 'Central st'),
  ('Chuck', 'Main Road'),
  ('Viola', 'Sideway')
]

mycursor.executemany(sql, val)

mydb.commit()

print(mycursor.rowcount, "was inserted.") 
```

### Ottieni l'ID inserito

Puoi ottenere l'Id della fila Basta inserire chiedendo l'oggetto cursore.

```python
sql = "INSERT INTO Utente (nome, cognome) VALUES (%s, %s)"
val = ("Michelle", "Blue Village")
mycursor.execute(sql, val)

mydb.commit()

print("1 record inserted, ID:", mycursor.lastrowid) 
```
